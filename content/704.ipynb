# 7.5. Propiedades de la transformada Z

En esta página, se muestran tanto pares básicos de la transformada Z como sus propiedades.

#### Pares básicos de transformadas Z

En la {ref}`tab:T7:TZ:pares_basicos` se muestran los pares básicos de transformadas más comunes.  A diferencia de la transformada de Fourier, cabe destacar que **cada par viene acompañado de su región de convergencia (ROC)**.

| :--- | :---: | :--- |

| $(r^{n} \sin\Omega_0 n)u[n]$ | $\displaystyle \frac{(r \sin \Omega_0)z^{-1}}{1-(2r \cos \Omega_0)z^{-1} + r^{2}z^{-2}}$ | $\left\|z \right\| >r$ |

#### Propiedades de la transformada Z

Respecto a las propiedades de la transformada Z, éstas son similares a las de la transformada de Fourier en tiempo discreto. Sin embargo, aparte de las diferencias relacionadas con la región de convergencia, hay alguna propiedad que presenta diferencias sustanciales. Antes de destacar algún ejemplo de ello, en la {ref}`tab:T7:TZ:propiedades` se muestran las propiedades de la transformada Z.

:::{table} Propiedades de la transformada Z
:name: tab:T7:TZ:propiedades

| Propiedad | Señal | Transformada | ROC |
| :--- | :--- | :--- | :--- |
|  | $\begin{array}{c}x[n]\\ x_1[n]\\ x_2[n]\end{array}$ | $\begin{array}{c}X(z)\\ X_1(z)\\ X_2(z)\end{array}$ | $\begin{array}{c}R\\ R_1\\ R_2\end{array}$ |
| **Linealidad** | $a x_1[n] + b x_2[n]$ | $a X_1(z) + b X_2(z)$ | $\text{Al menos } R_1 \bigcap R_2$ |
| **Desplazamiento temporal** | $x[n-n_0]$ | $z^{-n_0} X(z)$ | $R \pm \left\{ 0 \right\} \pm \left\{ \infty \right\}$ |
| **Escalado en** $z$ | $\begin{array}{c}e^{j\Omega_0 n} x[n]\\ z_{0}^{n} x[n]\\ a^{n} x[n]\end{array}$ | $\begin{array}{c}X(e^{j\Omega_0 n} z)\\ X\left( \frac{z}{z_0} \right)\\ X\left( a^{-1} z \right)\end{array}$ | $\begin{array}{c}R\\ \left\|z_{0} \right\| R\\ \left\|a \right\| R\end{array}$  |
| **Inversión temporal** | $x[-n]$ | $X\left( z^{-1} \right)$ | $R^{-1}$ |
| **Expansión temporal** | $x_{(k)}[n]$ | $X\left( z^{k} \right)$ | $R^{1/k}$ |
| **Conjugación** | $x^{*}[n]$ | $X^{*}\left( z^{*} \right)$ | $R$ |
| **Convolución** | $x_1 [n] * x_2 [n]$ | $X_1 (z) X_2 (z)$ | $\text{Al menos } R_1 \bigcap R_2$ |
| **Primera diferencia** | $x[n]-x[-n-1]$ | $\left( 1-z^{-1} \right) X(z)$ | $\text{Al menos } R \bigcap \left( \left\|z \right\| >0 \right)$ |
| **Acumulación** | $\sum_{k=-\infty}^{n} x[k]$ | $\frac{1}{1-z^{-1}} X(z)$ | $\text{Al menos } R \bigcap \left( \left\|z \right\| >1 \right)$ |
| **Diferenciación en** $z$ | $n x[n]$ | $-z \frac{dX(z)}{dz}$ | $R$ |
:::

**Teorema del valor inicial:**
Si una señal discreta $x[n]$ está definida solamente para $n \geq 0$, su valor en el instante inicial es el límite de su transformada Z en el infinito:

\begin{equation*}
 x[n] = 0 \text{ } \forall n < 0 \Rightarrow x[0] = \lim_{z \to \infty} X(z).
\end{equation*}

:::{note} Sobre el teorema del valor inicial

El teorema del valor inicial permite calcular el primer valor de una secuencia sin necesidad de calcular la transformada Z inversa.

El teorema se puede razonar a partir de la definición de la transformada Z, sabiendo que $x[n] = 0$  $\forall n < 0$:

\begin{equation*}
 X(z) = \sum_{n=0}^{\infty}x[n] z^{-n} = x[0] + x[1] z^{-1} + x[2] z^{-2} + ...
\end{equation*}

Si $z \to \infty$, el único término no nulo en la expresión es $x[0]$.

Para poder aplicar el teorema de valor inicial, la transformada Z tiene que existir y su ROC tiene que contener el infinito.

:::
Cabe realizar algunas aclaraciones respecto a la ROC de algunas propiedades:

- $\left\| a \right\| R$ hace referencia al conjunto de puntos $\left\{ \left\|a \right\| z \right\}$ para $z$ en $R$.

- $R^{-1}$ es el conjunto de puntos $z^{-1}$ donde $z$ está en $R$.

- $R^{1/k}$ es el conjunto de puntos $z^{1/k}$ donde $z$ está en $R$.

Como se ha indicado anteriormente, aunque las propiedades de la transformada Z sean similares a las de la transformada de Fourier en tiempo discreto, hay alguna propiedad con diferencias importantes.

Para ilustrar un ejemplo de propiedades distintas, usamos la propiedad de inversión en el tiempo, que se puede utilizar con frecuencia en ejercicios.

:::{warning}Propiedad de inversión temporal
Si la transformada de Fourier de una señal en tiempo discreto $x[n]$ es $X(\Omega)$, y su transformada Z es $X(z)$:
```{math}
\mathcal{F}\{x[-n]\} = X(-\Omega) \quad \Rightarrow \quad \mathcal{Z}\{x[-n]\} = X(z^{-1}),
```

siendo la región de convergencia el conjunto de puntos $z^{-1}$ en los que $z$ está en la región de convergencia de $X(z)$ (se invierte la región de convergencia).

Cabe destacar que la transformada Z de $x[-n]$ no es $X(-z)$, como ocurriría con la transformada de Fourier, sino $X(z^{-1})$.
:::
