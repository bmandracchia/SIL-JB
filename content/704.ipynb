# Propiedades de la transformada Z

En la {ref}`tab:T7:TZ:pares_basicos` se muestran los pares básicos de transformadas más comunes.  A diferencia de la transformada de Fourier, cabe destacar que **cada par viene acompañado de su región de convergencia (ROC)**.

:::{table} Pares básicos de la transformada Z
:name: tab:T7:TZ:pares_basicos

| Señal | Transformada | ROC |
| :--- | :--- | :--- |
| $\delta[n]$ | 1 | $z$ |
| $u[n]$ | $\frac{1}{1-z^{-1}}$ | $\left\|z \right\| >1$ |
| $-u[-n-1]$ | $\frac{1}{1-z^{-1}}$ | $\left\|z \right\| <1$  |
| $\delta[n-m]$ | $z^{-m}$ | $\left\{\begin{array}{c}z-\{0\} \text{ si } m>0\\ z-\{\infty\} \text{ si } m<0\end{array}\right\}$ |
| $\alpha^{n}u[n]$ | $\frac{1}{1-\alpha z^{-1}}$ | $\left\|z \right\| > \left\| \alpha \right\|$ |
| $-\alpha^{n}u[-n-1]$ | $\frac{1}{1-\alpha z^{-1}}$ | $\left\|z \right\| < \left\| \alpha \right\|$ |
| $n \alpha^{n}u[n]$ | $\frac{\alpha z^{-1}}{(1-\alpha z^{-1})^2}$ | $\left\|z \right\| > \left\| \alpha \right\|$ |
| $-n \alpha^{n}u[-n-1]$ | $\frac{\alpha z^{-1}}{(1-\alpha z^{-1})^2}$ | $\left\|z \right\| < \left\| \alpha \right\|$ |
| $(\cos\Omega_0 n)u[n]$ | $\frac{1-(\cos \Omega_0)z^{-1}}{1-(2 \cos \Omega_0)z^{-1} + z^{-2}}$ | $\left\|z \right\| >1$ |
| $(\sin\Omega_0 n)u[n]$ | $\frac{(\sin \Omega_0)z^{-1}}{1-(2 \cos \Omega_0)z^{-1} + z^{-2}}$ | $\left\|z \right\| >1$ |
| $(r^{n} \cos\Omega_0 n)u[n]$  | $\frac{1-(r \cos \Omega_0)z^{-1}}{1-(2r \cos \Omega_0)z^{-1} + r^{2}z^{-2}}$ | $\left\|z \right\| >r$ |
| $(r^{n} \sin\Omega_0 n)u[n]$ | $\frac{(r \sin \Omega_0)z^{-1}}{1-(2r \cos \Omega_0)z^{-1} + r^{2}z^{-2}}$ | $\left\|z \right\| >r$ |
:::

Respecto a las propiedades de la transformada Z, éstas son similares a las de la transformada de Fourier en tiempo discreto. Sin embargo, aparte de las diferencias relacionadas con la región de convergencia, hay alguna propiedad que presenta diferencias sustanciales. Antes de destacar algún ejemplo de ello, en la {ref}`tab:T7:TZ:propiedades` se muestran las propiedades de la transformada Z. Ilustramos un ejemplo de ello con la propiedad de inversión en el tiempo.

:::{table} Propiedades de la transformada Z
:name: tab:T7:TZ:propiedades

| Propiedad | Señal | Transformada | ROC |
| :--- | :--- | :--- | :--- |
|  | $\begin{array}{c}x[n]\\ x_1[n]\\ x_2[n]\end{array}$ | $\begin{array}{c}X(z)\\ X_1(z)\\ X_2(z)\end{array}$ | $\begin{array}{c}R\\ R_1\\ R_2\end{array}$ |
| **Linealidad** | $a x_1[n] + b x_2[n]$ | $a X_1(z) + b X_2(z)$ | $\text{Al menos } R_1 \bigcap R_2$ |
| **Desplazamiento temporal** | $x[n-n_0]$ | $z^{-n_0} X(z)$ | $R \pm \left\{ 0 \right\} \pm \left\{ \infty \right\}$ |
| **Escalado en** $z$ | $\begin{array}{c}e^{j\Omega_0 n} x[n]\\ z_{0}^{n} x[n]\\ a^{n} x[n]\end{array}$ | $\begin{array}{c}X(e^{j\Omega_0 n} z)\\ X\left( \frac{z}{z_0} \right)\\ X\left( a^{-1} z \right)\end{array}$ | $\begin{array}{c}R\\ \left\|z_{0} \right\| R\\ \left\|a \right\| R\end{array}$  |
| **Inversión temporal** | $x[-n]$ | $X\left( z^{-1} \right)$ | $R^{-1}$ |
| **Expansión temporal** | $x_{(k)}[n]$ | $X\left( z^{k} \right)$ | $R^{1/k}$ |
| **Conjugación** | $x^{*}[n]$ | $X^{*}\left( z^{*} \right)$ | $R$ |
| **Convolución** | $x_1 [n] * x_2 [n]$ | $X_1 (z) X_2 (z)$ | $\text{Al menos } R_1 \bigcap R_2$ |
|  | $-n \alpha^{n}u[-n-1]$ | $\frac{\alpha z^{-1}}{(1-\alpha z^{-1})^2}$ | $\left\|z \right\| < \left\| \alpha \right\|$ |
|  | $(\cos\Omega_0 n)u[n]$ | $\frac{1-(\cos \Omega_0)z^{-1}}{1-(2 \cos \Omega_0)z^{-1} + z^{-2}}$ | $\left\|z \right\| >1$ |
|  | $(\sin\Omega_0 n)u[n]$ | $\frac{(\sin \Omega_0)z^{-1}}{1-(2 \cos \Omega_0)z^{-1} + z^{-2}}$ | $\left\|z \right\| >1$ |
|  | $(r^{n} \cos\Omega_0 n)u[n]$  | $\frac{1-(r \cos \Omega_0)z^{-1}}{1-(2r \cos \Omega_0)z^{-1} + r^{2}z^{-2}}$ | $\left\|z \right\| >r$ |
|  | $(r^{n} \sin\Omega_0 n)u[n]$ | $\frac{(r \sin \Omega_0)z^{-1}}{1-(2r \cos \Omega_0)z^{-1} + r^{2}z^{-2}}$ | $\left\|z \right\| >r$ |
:::

Si la transformada de Fourier de una señal en tiempo discreto $x[n]$ es $X(\Omega)$, y su transformada Z es $X(z)$:

\begin{equation}
 \mathcal{F}\{x[-n]\} = X(-\Omega) \quad \Rightarrow \quad \mathcal{Z}\{x[-n]\} = X(z^{-1}),
\end{equation}

siendo la región de convergencia el conjunto de puntos $z^{-1}$ en los que $z$ está en la región de convergencia de $X(z)$ (se invierte la región de convergencia).
