# Tema 3

# Objetivos del tema 3

- Entender la utilidad del análisis de Fourier para la descripción de señales y de sistemas LTI.
- Conocer las diferentes variantes de análisis de Fourier en tiempo continuo o discreto y para señales de energía o de potencia.
- Entender el análisis de Fourier como una representación de señales en un espacio dual, resultado de proyectar las funciones de interés en una base funcional ortogonal.
- Entender el análisis de Fourier como una representación de señales en términos de su contenido frecuencial.
- Saber deducir las ecuaciones de análisis y de síntesis del desarrollo en serie de Fourier.
- Saber deducir las ecuaciones de análisis y de síntesis de la transformada de Fourier en tiempo continuo.
- Saber deducir las propiedades, y entender la dualidad existente entre ellas, para el desarrollo en serie de Fourier y transformada de Fourier en tiempo continuo.
- Comprender el concepto de autofunción de un sistema LTI, y entender que las exponenciales complejas son autofunciones de cualquier sistema LTI.
- Saber caracterizar sistemas LTI a partir de su función de transferencia.
- Aprender a resolver problemas relacionados con la caracterización de sistemas LTI mediante análisis de Fourier.
