# 1.2. Parámetros de interés de las señales

Vemos algunos parámetros de interés de las señales, tanto continuas como discretas.

## Valor medio


### Valor medio en un intervalo

El **valor medio** de una señal representa el promedio de sus valores dentro de un intervalo dado.

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

Para una señal continua $x(t)$ definida en el intervalo $t_1 \le t \le t_2$:

```{math}
\bar{x} \overset{\Delta}{=}
\frac{1}{t_2-t_1}
\int_{t_1}^{t_2} x(t)\,dt
```


Si el intervalo es simétrico respecto al origen,
$-T \le t \le T$:

```{math}
\bar{x} \overset{\Delta}{=}
\frac{1}{2T}
\int_{-T}^{T} x(t)\,dt
```

::::

:::{grid-item-card} Señales discretas

Para una señal discreta $x[n]$ definida en el intervalo
$n_1 \le n \le n_2$:

```{math}
\bar{x} \overset{\Delta}{=}
\frac{1}{n_2-n_1+1}
\sum_{n=n_1}^{n_2} x[n]
```

Si el intervalo es simétrico respecto al origen,
$-N \le n \le N$:

```{math}
\bar{x} \overset{\Delta}{=}
\frac{1}{2N+1}
\sum_{n=-N}^{N} x[n]
```

:::

:::::


### Valor medio total

El **valor medio total**, o simplemente **valor medio**, describe el promedio de una señal cuando se considera todo su dominio.

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

Para una señal continua $x(t)$:

```{math}
x_\text{AV} \overset{\Delta}{=}
\lim_{T\to\infty}
\frac{1}{T}
\int_{-T/2}^{T/2} x(t)\,dt
````

De forma equivalente:

```{math}
x_\text{AV} \overset{\Delta}{=}
\lim_{T\to\infty}
\frac{1}{2T}
\int_{-T}^{T} x(t)\,dt
```

::::

::::{grid-item-card} Señales discretas

Para una señal discreta $x[n]$:

```{math}
x_\text{AV} \overset{\Delta}{=}
\lim_{N\to\infty}
\frac{1}{2N+1}
\sum_{n=-N}^{N} x[n]
```

::::

:::::

### Valor medio para señales periódicas

Si la señal es **periódica**, su valor medio puede calcularse sobre un único período.

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

Para una señal continua periódica de período $T$:

```{math}
x_\text{AV} \overset{\Delta}{=}
\frac{1}{T}
\int_{\langle T\rangle} x(t)\,dt
```

::::

::::{grid-item-card} Señales discretas

Para una señal discreta periódica de período $N$:

```{math}
x_\text{AV} \overset{\Delta}{=}
\frac{1}{N}
\sum_{n\in\langle N\rangle} x[n]
```

::::

:::::

## Valor de pico

El **valor de pico** mide el valor máximo que alcanza el módulo de una señal.

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

```{math}
x_p \overset{\Delta}{=}
\max\{|x(t)|,\ t\in\mathbb{R}\}
```

::::

::::{grid-item-card} Señales discretas

```{math}
x_p \overset{\Delta}{=}
\max\{|x[n]|,\ n\in\mathbb{Z}\}
```

::::

:::::

## Potencia instantánea

Por analogía con señales que representan magnitudes físicas, se puede hablar de **potencia** y **energía** asociadas a una señal.

Como ejemplo, si una señal representa el voltaje $v(t)$ aplicado a una resistencia de valor $R$, la **potencia instantánea** disipada viene dada por:

```{figure} figures/T1/2_1_fig15
---
width: 60%
---
```

```{math}
p(t)=v(t)i(t)=\frac{1}{R}v^2(t)=Ri^2(t)
```

donde $v(t)$ es el voltaje e $i(t)$ la intensidad de corriente que circula por la resistencia.

A partir de esta expresión se generalizan las definiciones de potencia y energía para cualquier señal, independientemente de que represente o no una magnitud eléctrica.

Por tanto, la potencia es proporcional al cuadrado de la señal. De forma general:

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

```{math}
P_i(t) \overset{\Delta}{=}
|x(t)|^2
```

::::

::::{grid-item-card} Señales discretas

```{math}
P_i[n] \overset{\Delta}{=}
|x[n]|^2
```

::::

:::::

## Energía

:::{important .simple icon=false} Energía
La **energía** de una señal se define como la integral, o suma, de su potencia instantánea.
:::

### Energía en un intervalo de tiempo

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

En el intervalo $t_1\le t\le t_2$:

```{math}
E \overset{\Delta}{=}
\int_{t_1}^{t_2} |x(t)|^2\,dt
```

Para un intervalo simétrico, $-T\le t\le T$:

```{math}
E_T \overset{\Delta}{=}
\int_{-T}^{T} |x(t)|^2\,dt
```

::::

::::{grid-item-card} Señales discretas

En el intervalo $n_1\le n\le n_2$:

```{math}
E \overset{\Delta}{=}
\sum_{n=n_1}^{n_2} |x[n]|^2
```

Para un intervalo simétrico, $-N\le n\le N$:

```{math}
E_N \overset{\Delta}{=}
\sum_{n=-N}^{N} |x[n]|^2
```

::::

:::::

### Energía total

La **energía total** corresponde a la energía de la señal en un intervalo infinito.

::::{card} Señales continuas

```{math}
E_\infty \overset{\Delta}{=}
\lim_{T\to\infty}
\int_{-T}^{T} |x(t)|^2\,dt
=
\int_{-\infty}^{\infty} |x(t)|^2\,dt
```

::::

::::{card} Señales discretas

```{math}
E_\infty \overset{\Delta}{=}
\lim_{N\to\infty}
\sum_{n=-N}^{N} |x[n]|^2
=
\sum_{n=-\infty}^{\infty} |x[n]|^2
```

::::


:::{warning} Atención
Hay señales para las que esta integral o sumatorio no converge. Por ejemplo:

```{math}
x_1(t)=\frac{1}{\sqrt{t}},\qquad x_2(t)=C
```

```{math}
x_1[n]=\frac{1}{\sqrt{n}},\qquad x_2[n]=C
```

En estos casos, $E_\infty=\infty$. Por ello, nos interesa introducir otra medida relacionada: la **potencia media**.
:::

## Potencia media

:::{important .simple icon=false} Potencia media
La **potencia media** es el valor medio de la potencia instantánea.
:::

En general, la potencia media viene dada por:

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

```{math}
P_\infty \overset{\Delta}{=}
\lim_{T\to\infty}
\frac{1}{T}
\int_{-T/2}^{T/2} |x(t)|^2\,dt
```

De forma equivalente:

```{math}
P_\infty \overset{\Delta}{=}
\lim_{T\to\infty}
\frac{1}{2T}
\int_{-T}^{T} |x(t)|^2\,dt
```

::::

::::{grid-item-card} Señales discretas

```{math}
P_\infty \overset{\Delta}{=}
\lim_{N\to\infty}
\frac{1}{2N+1}
\sum_{n=-N}^{N} |x[n]|^2
```

::::

:::::

### Potencia media para señales periódicas

Si la señal es **periódica**, la potencia media puede calcularse sobre un único período.

:::::{grid} 1 1 2 2
:gutter: 3

::::{grid-item-card} Señales continuas

Para una señal continua periódica de período $T$:

```{math}
P_\infty \overset{\Delta}{=}
\frac{1}{T}
\int_{\langle T\rangle} |x(t)|^2\,dt
```

::::

::::{grid-item-card} Señales discretas

Para una señal discreta periódica de período $N$:

```{math}
P_\infty \overset{\Delta}{=}
\frac{1}{N}
\sum_{n\in\langle N\rangle} |x[n]|^2
```

::::

:::::


**Ejemplo**: $x(t)=2.$

  ```{figure} figures/T1/2_1_fig16 
  ---
  width: 60%
  ---

  ```

```{math}
P_i(t)=|x(t)|^2=4.
```
```{math}
E_\infty=\int_{-\infty}^\infty |x(t)|^2 dt=\int_{-\infty}^\infty 4 dt=\infty.
```
```{math}
P_\infty=\lim_{T\to\infty}\frac{1}{T}\int_{-T/2}^{T/2} P_i(t) dt=\lim_{T\to\infty}\frac{4T}{T}=4.
```



## Señales de energía y de potencia

Según las definiciones de energía y de potencia de una señal, se puede hablar de tres clases de señales:

  :::{important .simple icon=false} **Señales de energía**
  Son señales con **energía total finita**:

  ```{math}
  0<E_\infty<\infty.
  ```
  :::

  Tienen $P_\infty=0$:
  \begin{equation}
    P_\infty=\lim_{T\to\infty}\frac{1}{2T}\int_{-T}^{T} |x(t)|^2dt=\lim_{T\to\infty}\frac{E_T}{2T}=0.
  \end{equation}
  \begin{equation}
    P_\infty=\lim_{N\to\infty}\frac{1}{2N+1}\sum\limits_{n=-N}^{N} |x[n]|^2=\lim_{N\to\infty}\frac{E_N}{2N+1}=0.
  \end{equation}


  **Ejemplo**:

  ```{figure} figures/T1/2_1_fig17 
  ---
  width: 60%
  ---

  ```

  
  ```{math}
  E_\infty=1 \Rightarrow P_\infty=0.
  ```

  :::{important .simple icon=false}  **Señales de potencia**
  Son señales con **potencia media finita**:
  ```{math}
  0<P_\infty<\infty \quad \Rightarrow \quad E_\infty=\infty.
  ```
  :::
  Dado que tienen $P_\infty>0$, integrando (sumando) en un intervalo infinito, se obtiene $E_\infty=\infty$.

  **Ejemplos**:

  $x_1(t)=2.$\
  $x_2(t)$: cualquier señal periódica con valor de pico, $x_p$, finito.
  
  :::{important .simple icon=false}  **Señales con $E_\infty$ y $P_\infty$ infinitas**
  Son señales con **energía y potencia media infinitas**:
  ```{math}
  P_\infty=\infty \quad , \quad E_\infty=\infty.
  ```
  :::

  **Ejemplo**:
  $x(t)=t.$
  ```{math}
  E_\infty=\lim_{T\to\infty}\int_{-T}^{T}|t|^2 dt=\lim_{T\to\infty}\int_{-T}^{T}t^2 dt=\lim_{T\to\infty}\frac{\left.t^3\right|_{-T}^{T}}{3}=
    \lim_{T\to\infty}\frac{2T^3}{3}=\infty.
  ```
  ```{math}
  P_\infty=\lim_{T\to\infty}\frac{1}{2T}\int_{-T}^{T}|t|^2 dt=\lim_{T\to\infty}\frac{E_T}{2T}=\lim_{T\to\infty}\frac{2T^3}{6T}=\infty.
  ```

`````python
import numpy as np

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, CustomJS, Select, Div, Slider
from bokeh.layouts import column, row

from utils.plot_helpers import style_math_axes

output_notebook(verbose=False, hide_banner=True)

# ==============================================================================
# 1. EJE TEMPORAL
# ==============================================================================

t_min, t_max = -5.0, 5.0
num_points = 3000
t = np.linspace(t_min, t_max, num_points)

initial_signal = "Senoidal"
A0 = 1.0
T0 = 2.0

# ==============================================================================
# 2. FUNCIÓN AUXILIAR
# ==============================================================================

def compute_signal(name, A, T):

    if name == "Senoidal":
        x = A*np.sin(2*np.pi*t/T)

    elif name == "Cosenoidal con offset":
        x = 0.5*A + A*np.cos(2*np.pi*t/T)

    elif name == "Pulso rectangular":
        x = A*((t >= -T/2) & (t <= T/2)).astype(float)

    elif name == "Exponencial decreciente":
        x = A*np.exp(-t)*(t >= 0)

    elif name == "Gaussiana":
        x = A*np.exp(-t**2)

    elif name == "Señal cuadrada periódica":
        x = A*np.sign(np.sin(2*np.pi*t/T))

    return x

# ==============================================================================
# 3. ESTADO INICIAL
# ==============================================================================

x0 = compute_signal(initial_signal, A0, T0)

p_inst0 = x0**2

mean0 = np.trapezoid(x0, t)/(t_max - t_min)

energy0 = np.trapezoid(p_inst0, t)

peak0 = np.max(np.abs(x0))

if initial_signal == "Senoidal":
    power0 = A0**2/2

elif initial_signal == "Cosenoidal con offset":
    power0 = A0**2/4 + A0**2/2

elif initial_signal == "Señal cuadrada periódica":
    power0 = A0**2

else:
    power0 = 0

source = ColumnDataSource(data=dict(
    t=t,
    x=x0,
    p_inst=p_inst0,
    mean=np.full_like(t, mean0),
    peak_pos=np.full_like(t, peak0),
    p_mean=np.full_like(t, power0),
))

# ==============================================================================
# 4. FIGURA 1: SEÑAL
# ==============================================================================

p1 = figure(
    height=350,
    width=600,
    title="Señal x(t)",
    tools="pan,wheel_zoom,reset"
)

p1.line(
    "t", "x",
    source=source,
    line_width=3,
    color="navy",
    legend_label="x(t)"
)

p1.line(
    "t", "mean",
    source=source,
    line_width=2,
    color="firebrick",
    line_dash="dashed",
    legend_label="valor medio en \nel intervalo mostrado"
)

p1.line(
    "t", "peak_pos",
    source=source,
    line_width=2,
    color="darkorange",
    line_dash="dotdash",
    legend_label="valor de pico"
)

p1.line(
    t,
    np.zeros_like(t),
    color="black",
    alpha=0.4
)

p1.legend.location = "top_right"
p1.legend.click_policy = "hide"

p1.grid.grid_line_alpha = 0.25

style_math_axes(
    p1,
    x_range=(t_min, t_max),
    y_range=(-2.5, 2.5),
    xlabel="t",
    ylabel="x(t)"
)

# ==============================================================================
# 5. FIGURA 2: POTENCIA INSTANTÁNEA
# ==============================================================================

p2 = figure(
    height=320,
    width=600,
    title="Potencia instantánea p(t) = |x(t)|²",
    tools="pan,wheel_zoom,reset"
)

p2.varea(
    x="t",
    y1=0,
    y2="p_inst",
    source=source,
    alpha=0.25,
    color="seagreen",
    legend_label="energía = área bajo p(t)"
)

p2.line(
    "t", "p_inst",
    source=source,
    line_width=3,
    color="seagreen",
    legend_label="p(t) = |x(t)|²"
)

p2.line(
    "t", "p_mean",
    source=source,
    line_width=2,
    color="firebrick",
    line_dash="dashed",
    legend_label="potencia media total"
)

p2.line(
    t,
    np.zeros_like(t),
    color="black",
    alpha=0.4
)

p2.legend.location = "top_right"
p2.legend.click_policy = "hide"

p2.grid.grid_line_alpha = 0.25

style_math_axes(
    p2,
    x_range=(t_min, t_max),
    y_range=(0, 4.5),
    xlabel="t",
    ylabel="p(t)"
)

# ==============================================================================
# 6. CONTROLES
# ==============================================================================

select = Select(
    title="Tipo de señal",
    value=initial_signal,
    options=[
        "Senoidal",
        "Cosenoidal con offset",
        "Pulso rectangular",
        "Exponencial decreciente",
        "Gaussiana",
        "Señal cuadrada periódica",
    ],
    width=250
)

amp_slider = Slider(
    title="Amplitud A",
    start=0.2,
    end=2.0,
    value=A0,
    step=0.1,
    width=150
)

period_slider = Slider(
    title="Período / anchura T",
    start=0.5,
    end=5.0,
    value=T0,
    step=0.1,
    width=150
)

# ==============================================================================
# 7. TEXTO INFORMATIVO
# ==============================================================================

info = Div(width=600)

info.text = f"""
<b>Parámetros calculados</b><br>
Valor medio en el intervalo mostrado: {mean0:.3f}<br>
Valor de pico: {peak0:.3f}<br>
Energía en el intervalo mostrado: {energy0:.3f}<br>
Potencia media total: {power0:.3f}
"""

# ==============================================================================
# 8. CALLBACK
# ==============================================================================

callback = CustomJS(
    args=dict(
        source=source,
        select=select,
        amp_slider=amp_slider,
        period_slider=period_slider,
        info=info,
        t_min=t_min,
        t_max=t_max
    ),
    code="""

    const data = source.data;
    const t = data["t"];

    const name = select.value;

    const A = amp_slider.value;
    const T = period_slider.value;

    const x = [];
    const p_inst = [];

    for (let i = 0; i < t.length; i++) {

        const ti = t[i];

        let value = 0;

        // ==========================================================
        // DEFINICIÓN DE LA SEÑAL
        // ==========================================================

        if (name === "Senoidal") {

            value = A*Math.sin(2*Math.PI*ti/T);

        }

        else if (name === "Cosenoidal con offset") {

            value = 0.5*A + A*Math.cos(2*Math.PI*ti/T);

        }

        else if (name === "Pulso rectangular") {

            value = (ti >= -T/2 && ti <= T/2) ? A : 0;

        }

        else if (name === "Exponencial decreciente") {

            value = (ti >= 0) ? A*Math.exp(-ti) : 0;

        }

        else if (name === "Gaussiana") {

            value = A*Math.exp(-ti*ti);

        }

        else if (name === "Señal cuadrada periódica") {

            value = A*Math.sign(Math.sin(2*Math.PI*ti/T));

        }

        x.push(value);

        p_inst.push(value*value);
    }

    // ==============================================================
    // VALOR MEDIO Y ENERGÍA (TRAPECIOS)
    // ==============================================================

    let integral_x = 0;
    let energy = 0;

    for (let i = 0; i < t.length - 1; i++) {

        const dt = t[i+1] - t[i];

        integral_x += 0.5*(x[i] + x[i+1])*dt;

        energy += 0.5*(p_inst[i] + p_inst[i+1])*dt;
    }

    const duration = t_max - t_min;

    const mean = integral_x/duration;

    // ==============================================================
    // POTENCIA MEDIA TOTAL
    // ==============================================================

    let power_mean_total = 0;

    if (name === "Senoidal") {

        power_mean_total = A*A/2;

    }

    else if (name === "Cosenoidal con offset") {

        power_mean_total = A*A/4 + A*A/2;

    }

    else if (name === "Señal cuadrada periódica") {

        power_mean_total = A*A;

    }

    else {

        power_mean_total = 0;
    }

    // ==============================================================
    // VALOR DE PICO
    // ==============================================================

    let peak = 0;

    for (let i = 0; i < x.length; i++) {

        peak = Math.max(peak, Math.abs(x[i]));
    }

    // ==============================================================
    // CLASIFICACIÓN
    // ==============================================================

    let classification = "";

    if (
        name === "Senoidal" ||
        name === "Cosenoidal con offset" ||
        name === "Señal cuadrada periódica"
    ) {

        classification =
        "Es una señal de potencia: su energía total tiende a infinito, pero su potencia media es finita.";
    }

    else {

        classification =
        "Es una señal de energía: su energía total es finita y su potencia media total tiende a cero.";
    }

    // ==============================================================
    // ACTUALIZAR DATOS
    // ==============================================================

    data["x"] = x;

    data["p_inst"] = p_inst;

    data["mean"] = Array(t.length).fill(mean);

    data["peak_pos"] = Array(t.length).fill(peak);

    data["p_mean"] = Array(t.length).fill(power_mean_total);

    // ==============================================================
    // ACTUALIZAR TEXTO
    // ==============================================================

    info.text = `
    <b>Parámetros calculados</b><br>
    Valor medio en el intervalo mostrado: ${mean.toFixed(3)}<br>
    Valor de pico: ${peak.toFixed(3)}<br>
    Energía en el intervalo mostrado: ${energy.toFixed(3)}<br>
    Potencia media total: ${power_mean_total.toFixed(3)}<br><br>
    ${classification}
    `;

    source.change.emit();

"""
)

select.js_on_change("value", callback)

amp_slider.js_on_change("value", callback)

period_slider.js_on_change("value", callback)

# ==============================================================================
# 9. LAYOUT
# ==============================================================================

layout = column(
    p1,
    p2,
    row(select, amp_slider, period_slider),
    info
)

show(layout)
`````
