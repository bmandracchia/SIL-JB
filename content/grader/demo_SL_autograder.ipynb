## 25. Ajuste interactivo de la amplitud de una senoide

La línea discontinua representa una señal de referencia. Ajusta la amplitud $A$ de

$$
x(t)=A\sin(2\pi t)
$$

hasta que la señal continua coincida con la referencia. Cuando estés satisfecho con el ajuste, pulsa **Guardar respuesta**.


```python
t = np.linspace(0, 2, 500)
A_ref = 2.3
x_ref = A_ref * np.sin(2 * np.pi * t)

amplitude = widgets.FloatSlider(
    value=1.0,
    min=0.0,
    max=4.0,
    step=0.1,
    description="Amplitud:",
    continuous_update=True,
    style={"description_width": "initial"},
    layout=widgets.Layout(width="520px"),
)

save_button = widgets.Button(
    description="Guardar respuesta",
    button_style="primary",
)
message = widgets.HTML()
plot_output = widgets.Output()

def update_plot(change=None):
    A = amplitude.value
    x = A * np.sin(2 * np.pi * t)

    with plot_output:
        plot_output.clear_output(wait=True)
        plt.figure(figsize=(8, 4))
        plt.plot(t, x_ref, "--", label="Señal de referencia")
        plt.plot(t, x, label="Tu señal")
        plt.xlabel("Tiempo $t$")
        plt.ylabel("$x(t)$")
        plt.grid(True)
        plt.legend()
        plt.show()

def save_amplitude_answer(_):
    set_answer(
        "q25",
        {"amplitude": amplitude.value},
        question_type="parameters",
        complete=True,
    )
    message.value = (
        "<span style='color:#2e7d32'><b>Respuesta guardada:</b> "
        f"A = {amplitude.value:.2f}</span>"
    )

amplitude.observe(update_plot, names="value")
save_button.on_click(save_amplitude_answer)

update_plot()
display(widgets.VBox([amplitude, plot_output, save_button, message]))
```
