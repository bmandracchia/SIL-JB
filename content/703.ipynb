# La transformada Z inversa


La transformada Z inversa se puede deducir teniendo en cuenta que, de acuerdo con la ecuación de análisis, es una serie de Laurent en $z$ y una función holomorfa dentro de su región de convergencia.

En cambio, de manera más sencilla, se va a deducir la expresión de la transformada Z inversa considerado que la transformada Z se puede expresar en función de la transformada de Fourier:

\begin{equation}
 X(z) = \mathcal{F} \{x[n]r^{-n} \} , \quad z = r e^{j \Omega}
\end{equation}

si $r$ es un valor real positivo que pertenece a la región de convergencia. Por lo tanto:

\begin{equation}
 x[n]r^{-n} = \mathcal{F}^{-1} \{X(z) \}.
\end{equation}

Se obtiene la transformada Z inversa a partir de la ecuación de síntesis de la transformada de Fourier en tiempo discreto:

\begin{equation}
 x[n]r^{-n} = \frac{1}{2\pi} \int_{-\pi}^{\pi} X\left( re^{j\Omega} \right) e^{j\Omega n} d\Omega \\\Rightarrow x[n] = \frac{1}{2\pi} \int_{-\pi}^{\pi} X\left( re^{j\Omega} \right) r^{n} e^{j\Omega n} d\Omega = \frac{1}{2\pi} \int_{-\pi}^{\pi} X\left( re^{j\Omega} \right) \left( re^{j\Omega}\right)^{n} d\Omega.
\end{equation}

Para expresar este resultado en función de $z$, realizamos un cambio de variable:

\begin{equation}
 z = re^{j\Omega}, \quad dz = jre^{j\Omega}d\Omega=jzd\Omega.
\end{equation}

Lo que da lugar a la ecuación de síntesis de la transformada Z.

:::{note .simple icon=false} Ecuación de síntesis de la transformada Z

\begin{equation*}
 x[n] = \frac{1}{2\pi j} \oint_r X(z) z^{n-1}dz
\end{equation*}

siendo $\oint_r$ la integral en un contorno cerrado (cualquier curva) en sentido antihorario, centrado en el origen y con radio $r$ perteneciente a la región de convergencia.

:::

En el contexto de la asignatura, recurriremos a otras técnicas como la descomposición en fracciones simples, las propiedades de la transformada Z, y los pares conocidos de transformadas Z [disponibles](#sec:T7:TZ:propiedades).

:::{warning .simple icon=false} Deducción de la ecuación de síntesis sin usar la transformada de Fourier

El cálculo de la transformada Z inversa es un problema equiparable a la extracción de coeficientes en series de Laurent, siendo $x[n]$ el coeficiente de $z^{-n}$, usando integrales complejas.

Multiplicando a ambos lados de la ecuación de análisis por $z^{n-1}$ y poniendo como índice del sumatorio $k$ en lugar de $n$:

\begin{equation*}
 X(z) \cdot z^{n-1} = \sum_{k=-\infty}^{\infty} x[k] \cdot z^{n-1-k}
\end{equation*}

\begin{equation*}
 \oint_C X(z) \cdot z^{n-1} dz = \sum_{k=-\infty}^{\infty} x[k] \cdot \oint_C z^{n-1-k} dz
\end{equation*}

siendo $\oint_C$ la integral sobre una curva cerrada $C$. El valor de la segunda integral es $2\pi j$ si $n=k$, y 0 en cualquier otro caso.[^1]

[^1]: Para realizar el cálculo de la integral, se puede usar el cambio de variable $dz=jzd\Omega$ y aplicar el teorema integral de Cauchy.

Por lo tanto, sustituyendo la segunda integral por $2\pi j$, quitando el sumatorio, sustituyendo $x[k]$ por $x[n]$, y dividiendo a ambos lados de la ecuación por $2\pi j$, el resultado es la ecuación de síntesis.

:::
