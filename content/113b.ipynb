# 1.4.2. Señales impulso unitario y escalón unitario

A continuación se presentan otras señales básicas como son el impulso unitario o el escalón unitario, tanto continuo como discreto. Estas señales serán fundamentales para la caracterización de señales y sistemas que veremos en el tema 3. Comenzaremos por las señales discretas, que son más sencillas en este caso que las continuas.

## Discretas

### Impulso unitario
:::{important .simple icon=false} Impulso unitario[^6]: $\delta[n]$
[^6]: También llamado delta de Kronecker, especialmente en matemáticas.

\begin{equation}
  \delta[n]=
    \begin{cases}
0, & n\neq 0,\\
1, & n=0.
    \end{cases}
\end{equation}
:::

Es una señal que vale 1 exclusivamente cuando su argumento es cero y cero en caso contrario.

### Escalón unitario
:::{important .simple icon=false} Escalón unitario: $u[n]$

\begin{equation}
  u[n]=
    \begin{cases}
0, & n<0,\\
1, & n\ge 0.
    \end{cases}
\end{equation}
:::

Es una señal que vale 1 cuando su argumento es mayor o igual que cero y cero en caso contrario.

### Relación entre Impulso y Escalón 
Se puede ver que ambas señales están relacionadas. 

:::{tip .simple} El impulso unitario discreto es la primera diferencia del escalón unitario discreto
\begin{equation}
  \delta[n]=u[n]-u[n-1].
\end{equation}
::: 

```{figure} figures/T1/2_4_fig3 
---
width: 60%
---
```


 
Por su parte:

:::{tip .simple} El escalón unitario es la sumación del impulso unitario
\begin{equation}
  u[n]=\sum_{m=-\infty}^n \delta[m].
\end{equation}
:::

```{figure} figures/T1/2_4_fig4 
---
width: 75%
---
```


Haciendo un cambio de variable, esta expresión se transforma en otra equivalente que será útil en el tema 3:
```{math}
\text{c.v.: }\ k=n-m\ \Rightarrow\ m=n-k,
```
```{math}
u[n]=\sum_{k=\infty}^0\delta[n-k],
```
y colocando el extremo menor del sumatorio abajo, como es norma habitual:
\begin{equation}
  u[n]=\sum_{k=0}^\infty\delta[n-k].
\end{equation}

```{figure} figures/T1/2_4_fig5 
---
width: 75%
---
```


La última expresión es equivalente a la suma de impulsos retardados (usado en el tema 3):
\begin{equation}
  u[n]=\sum_{k=0}^\infty\delta[n-k]=...+\delta[n+2]+\delta[n+1]+\delta[n]+\delta[n-1]+\delta[n-2]+....
\end{equation}


### Propiedad de muestreo del impulso unitario discreto
El impulso unitario discreto se puede usar para obtener muestras de una señal discreta:

:::{important .simple icon=false} Por ejemplo, en $n=0$:
  \begin{equation}\label{prop_muestro_delta_discreta}

  \end{equation}
:::

### Escalón unitario
Empezamos en este caso definiendo el escalón unitario.

:::{important .simple icon=false} Escalón unitario: $u(t)$

Se define de forma similar al escalón unitario discreto:
\begin{equation}
  u(t)=
    \begin{cases}
0, & t<0,\\
1, & t>0.
    \end{cases}
\end{equation}
:::

```{figure} figures/T1/2_4_fig9 
---
width: 60%
---
```

Se puede ver que es discontinuo en $t=0$, por lo que la definición del impulso unitario no será tan sencilla como en el caso discreto, si bien su relación sí será su análoga, como veremos a continuación.



### Impulso unitario: $\delta(t)$

Partimos de relaciones con el escalón unitario equivalentes al caso discreto. En lugar de una sumación aparece una integración y, en lugar de una primera diferencia, una derivación:

\begin{equation}\label{escalon_continuo}
u(t)=\int_{-\infty}^{t}\delta(\tau)\,d\tau,
\end{equation}

\begin{equation}
\delta(t)=\frac{du(t)}{dt}.
\end{equation}

:::{warning} Atención
La última expresión presenta problemas formales, ya que $u(t)$ **no es continua en** $t=0$ y, por tanto, no es derivable en el sentido clásico.
:::

Para interpretar esta relación resulta útil aproximar la delta de Dirac mediante una familia de funciones suaves y derivables. Una elección especialmente conveniente es la **gaussiana normalizada**

\begin{equation}
\delta_\sigma(t)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-t^2/(2\sigma^2)},
\end{equation}

donde $\sigma$ controla su anchura.

A medida que $\sigma$ disminuye, la gaussiana se concentra alrededor del origen. Para cualquier $t\neq0$, su valor tiende a cero, mientras que su altura en $t=0$ aumenta sin límite. Sin embargo, su área permanece constante:

\begin{equation}
\int_{-\infty}^{\infty}\delta_\sigma(t)\,dt=1.
\end{equation}





```{admonition} Exploración interactiva
:class: tip

La siguiente demostración interactiva permite comprobar este comportamiento modificando el valor de $\sigma$. También permite visualizar las dos primeras derivadas de la aproximación gaussiana, que convergen hacia las distribuciones $\delta'(t)$ y $\delta''(t)$.

```


````python
import numpy as np

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import (
    ColumnDataSource,
    CustomJS,
    Slider,
    Select,
    Div,
    Span,
)
from bokeh.layouts import column, row

from utils.plot_helpers import style_math_axes

output_notebook(verbose=False, hide_banner=True)

# ==============================================================================
# 1. EJE TEMPORAL
# ==============================================================================

t_min, t_max = -3.0, 3.0
num_points = 4001
t = np.linspace(t_min, t_max, num_points)

sigma0 = 0.5
initial_case = "Aproximación de δ(t)"

# ==============================================================================
# 2. APROXIMACIONES GAUSSIANAS
# ==============================================================================

def gaussian_delta(t, sigma):
    return (
        np.exp(-t**2 / (2 * sigma**2))
        / (sigma * np.sqrt(2 * np.pi))
    )


def gaussian_delta_first(t, sigma):
    delta = gaussian_delta(t, sigma)
    return -(t / sigma**2) * delta


def gaussian_delta_second(t, sigma):
    delta = gaussian_delta(t, sigma)
    return ((t**2 / sigma**4) - (1 / sigma**2)) * delta


y0 = gaussian_delta(t, sigma0)

source = ColumnDataSource(
    data=dict(
        t=t,
        y=y0,
        zero=np.zeros_like(t),
    )
)

# ==============================================================================
# 3. FIGURA
# ==============================================================================

p = figure(
    height=440,
    width=680,
    title="Aproximación gaussiana de la delta de Dirac",
    tools="pan,wheel_zoom,box_zoom,reset"
)

p.line(
    "t",
    "y",
    source=source,
    line_width=3,
    color="navy",
    legend_label="δσ(t)"
)

p.line(
    "t",
    "zero",
    source=source,
    color="black",
    alpha=0.4
)

origin = Span(
    location=0,
    dimension="height",
    line_color="black",
    line_dash="dotted",
    line_alpha=0.4
)

p.add_layout(origin)

p.legend.location = "top_right"
p.grid.grid_line_alpha = 0.25

style_math_axes(
    p,
    x_range=(t_min, t_max),
    y_range=(-5, 5),
    xlabel="t",
    ylabel="amplitud"
)

# ==============================================================================
# 4. CONTROLES
# ==============================================================================

case_select = Select(
    title="Función representada",
    value=initial_case,
    options=[
        "Aproximación de δ(t)",
        "Primera derivada δ′(t)",
        "Segunda derivada δ″(t)",
    ],
    width=230
)

sigma_slider = Slider(
    title="Anchura σ",
    start=0.01,
    end=1.0,
    value=sigma0,
    step=0.01,
    width=260
)

info = Div(width=680)

area0 = np.trapezoid(y0, t)
peak0 = gaussian_delta(np.array([0.0]), sigma0)[0]

info.text = f"""
<div style="
    padding: 10px;
    border-left: 5px solid navy;
    background-color: #f2f5fa;
">
    <b>Aproximación gaussiana de la delta de Dirac</b><br><br>

    <span style="font-size:18px;">
        δ<sub>σ</sub>(t)
        =
        1/(σ√(2π))
        exp[-t²/(2σ²)]
    </span><br><br>

    σ = {sigma0:.2f}<br>
    Altura en t = 0: <b>{peak0:.4f}</b><br>
    Área numérica: <b>{area0:.6f}</b><br><br>

    Al disminuir σ, la función se concentra alrededor de t = 0,
    aumenta su altura y conserva un área igual a 1.
</div>
"""

# ==============================================================================
# 5. CALLBACK
# ==============================================================================

callback = CustomJS(
    args=dict(
        source=source,
        case_select=case_select,
        sigma_slider=sigma_slider,
        info=info,
        plot=p
    ),
    code="""
    const data = source.data;
    const t = data["t"];

    const name = case_select.value;
    const sigma = sigma_slider.value;

    const y = [];

    const dt = t[1] - t[0];

    let area = 0;
    let firstMoment = 0;
    let absoluteArea = 0;
    let maxAbs = 0;

    function gaussianDelta(time) {
        return (
            Math.exp(
                -(time * time) / (2 * sigma * sigma)
            )
            / (sigma * Math.sqrt(2 * Math.PI))
        );
    }

    function functionValue(time) {
        const delta = gaussianDelta(time);

        if (name === "Aproximación de δ(t)") {
            return delta;
        }

        if (name === "Primera derivada δ′(t)") {
            return -(time / (sigma * sigma)) * delta;
        }

        if (name === "Segunda derivada δ″(t)") {
            return (
                (time * time) / Math.pow(sigma, 4)
                - 1 / (sigma * sigma)
            ) * delta;
        }

        return 0;
    }

    for (let i = 0; i < t.length; i++) {
        const value = functionValue(t[i]);

        y.push(value);
        maxAbs = Math.max(maxAbs, Math.abs(value));
    }

    /*
    Integración numérica mediante la regla del trapecio.
    */
    for (let i = 0; i < t.length - 1; i++) {
        const y1 = y[i];
        const y2 = y[i + 1];

        const t1 = t[i];
        const t2 = t[i + 1];

        area += 0.5 * (y1 + y2) * dt;

        firstMoment += (
            0.5 * (t1 * y1 + t2 * y2) * dt
        );

        absoluteArea += (
            0.5 * (Math.abs(y1) + Math.abs(y2)) * dt
        );
    }

    data["y"] = y;

    /*
    Ajuste automático del eje vertical para evitar que los picos
    desaparezcan cuando sigma disminuye.
    */
    const margin = 1.15 * maxAbs;

    if (name === "Aproximación de δ(t)") {
        plot.y_range.start = -0.05 * margin;
        plot.y_range.end = margin;
    }
    else {
        plot.y_range.start = -margin;
        plot.y_range.end = margin;
    }

    const peak = gaussianDelta(0);

    if (name === "Aproximación de δ(t)") {
        info.text = `
        <div style="
            padding: 10px;
            border-left: 5px solid navy;
            background-color: #f2f5fa;
        ">
            <b>Aproximación gaussiana de la delta de Dirac</b><br><br>

            <span style="font-size:18px;">
                δ<sub>σ</sub>(t)
                =
                1/(σ√(2π))
                exp[-t²/(2σ²)]
            </span><br><br>

            σ = ${sigma.toFixed(2)}<br>
            Altura en t = 0:
            <b>${peak.toFixed(4)}</b><br>

            Área numérica:
            <b>${area.toFixed(6)}</b><br><br>

            Cuando σ disminuye, la gaussiana se estrecha y aumenta
            su altura, pero su masa total permanece igual a 1:

            <br><br>

            <span style="font-size:18px;">
                ∫ δ<sub>σ</sub>(t) dt = 1.
            </span>
        </div>
        `;
    }

    else if (name === "Primera derivada δ′(t)") {
        info.text = `
        <div style="
            padding: 10px;
            border-left: 5px solid firebrick;
            background-color: #fff3f1;
        ">
            <b>Primera derivada de la aproximación gaussiana</b><br><br>

            <span style="font-size:18px;">
                δ′<sub>σ</sub>(t)
                =
                -t/σ² · δ<sub>σ</sub>(t)
            </span><br><br>

            σ = ${sigma.toFixed(2)}<br>
            Integral numérica:
            <b>${area.toExponential(3)}</b><br>

            Primer momento:
            <b>${firstMoment.toFixed(6)}</b><br><br>

            La función es impar y su integral es nula. En el límite,
            representa la derivada de la delta de Dirac:

            <br><br>

            <span style="font-size:18px;">
                ∫ δ′(t)φ(t)dt = −φ′(0).
            </span>
        </div>
        `;
    }

    else {
        info.text = `
        <div style="
            padding: 10px;
            border-left: 5px solid darkorange;
            background-color: #fff8eb;
        ">
            <b>Segunda derivada de la aproximación gaussiana</b><br><br>

            <span style="font-size:18px;">
                δ″<sub>σ</sub>(t)
                =
                (t²/σ⁴ − 1/σ²)δ<sub>σ</sub>(t)
            </span><br><br>

            σ = ${sigma.toFixed(2)}<br>
            Integral numérica:
            <b>${area.toExponential(3)}</b><br>

            Área absoluta:
            <b>${absoluteArea.toFixed(4)}</b><br><br>

            La función es par y presenta un lóbulo central negativo
            acompañado por dos lóbulos positivos. En el límite:

            <br><br>

            <span style="font-size:18px;">
                ∫ δ″(t)φ(t)dt = φ″(0).
            </span>
        </div>
        `;
    }

    source.change.emit();
    """
)

case_select.js_on_change("value", callback)
sigma_slider.js_on_change("value", callback)

# ==============================================================================
# 6. LAYOUT
# ==============================================================================

layout = column(
    p,
    row(case_select, sigma_slider),
    info
)

show(layout)
````


En el límite,

\begin{equation}
\delta(t)=
\lim_{\sigma\to0}
\delta_\sigma(t).
\end{equation}

:::{important .simple icon=false} Definición
La delta de Dirac $\delta(t)$ es una distribución idealizada cuya masa está concentrada en el instante $t=0$. Aunque no puede interpretarse como una función ordinaria, satisface

\begin{equation}
\int_{-\infty}^{\infty}\delta(t)\,dt=1.
\end{equation}
:::

Se representa mediante una flecha situada en $t=0$, indicando que toda su área se encuentra concentrada en dicho instante.

```{figure} figures/T1/2_4_fig12
---
width: 60%
---
```

:::{warning} Atención
La altura de la flecha y la etiqueta que la acompaña representan el **área** del impulso, no su altura, ya que $\delta(t)$ es una señal de área y no una función con amplitud finita.
:::

Un impulso escalado, $C\delta(t)$, tiene área $C$:

\begin{equation}
\int_{-\infty}^{\infty}C\delta(t)\,dt=C,
\end{equation}

\begin{equation}
\int_{-\infty}^{t}C\delta(\tau)\,d\tau
=
Cu(t).
\end{equation}

### Relación entre Impulso y Escalón 

Podemos realizar las mismas interpretaciones gráficas que en el caso discreto, haciendo los cambios oportunos del mundo discreto al continuo. Así la ecuación [](#escalon_continuo):
```{math}
u(t)=\int_{-\infty}^t\delta(\tau)d\tau,
```

```{figure} figures/T1/2_4_fig14 
---
width: 75%
---
```



Haciendo el cambio de variable $\sigma=t-\tau$ a  $\tau=t-\sigma$:
\begin{equation}
  u(t)=\int_{-\infty}^t\delta(\tau)d\tau=\int_\infty^0\delta(t-\sigma)(-d\sigma)=\int_0^\infty\delta(t-\sigma)d\sigma.
\end{equation}
Teniendo en cuenta que el impulso unitario es una señal par:
\begin{equation}
  \delta(t-\sigma)=\delta(\sigma-t),
\end{equation}

### Propiedad de muestreo del impulso unitario continuo
Al igual que ocurría en el caso discreto, el impulso unitario continuo se puede emplear para obtener muestras de una señal continua (esto será muy importante en los temas 3 y 7).

Lo vemos mediante la aproximación $\delta_\Delta(t)$:

```{figure} figures/T1/2_4_fig16 
---
width: 60%
---
```


\begin{equation}
  x_1(t)=x(t)\delta_\Delta(t).
\end{equation}
Si $\Delta\ll\ \Rightarrow\ x(t)$ es aproximadamente constante en $0\le t\le\Delta$:
\begin{equation}
  x(t)\delta_\Delta(t)\simeq x(0)\delta_\Delta(t).
\end{equation}

:::{important .simple icon=false}  En el límite cuando $\Delta\to0$:
  \begin{equation}\label{prop_muestro_delta_continua}
    x(t)\delta(t)=x(0)\delta(t).
  \end{equation}
:::

De forma similar, para un impulso unitario desplazado:
  \begin{equation}
    x(t)\delta(t-t_0)=x(t_0)\delta(t-t_0).
  \end{equation}

A la ecuación [](#prop_muestro_delta_continua) se la denomina **propiedad de muestreo del impulso unitario continuo**.
