# 3.2.6. Interpretación espectral de la serie de Fourier

Desde un punto de vista espectral, la serie de Fourier proporciona una representación discreta del contenido en frecuencia de una señal periódica. Cada coeficiente $c_k$ está asociado a una componente armónica de frecuencia $k\omega_0$, de modo que la periodicidad en el dominio temporal se traduce en una discretización del espectro en el dominio de la frecuencia. Como ejemplo, se muestra en la figura la serie de Fourier correspondiente a un tren de pulsos triangulares.

```{figure} figures/T3/signal_tri_periodic2.svg
:name: figs:T3:SF:interpretacion_tiempo
:width: 80%
:alt: Tren de pulsos triangulares.
:align: center
```
```{figure} figures/T3/signal_tri_periodic2_dk.svg
:name: figs:T3:SF:interpretacion_frecuencia
:width: 80%
:alt: Serie de Fourier correspondiente a un tren de pulsos triangulares.
:align: center
```

Esta interpretación permite analizar de forma cualitativa el comportamiento de las señales y los sistemas:
- Señales cuyos coeficientes son significativos únicamente para valores pequeños de $|k|$ presentan una variación lenta en el tiempo.
- La presencia de armónicos de orden elevado se asocia a variaciones rápidas o a la existencia de discontinuidades.

:::{tip} Interpretación espectral: Suavidad / Abruptidad
En la siguiente gráfica interactiva se analiza la relación entre la forma de onda de la señal $\tilde{x}(t)$ y el módulo de los coeficientes de su serie de Fourier, $|c_k|$, siendo:
```{math}
    \tilde{x}(t) \propto \tanh(s \cdot (\sin(\omega_0 t) + a))
```
Variando los parámetros se puede pasar de una señal puramente senoidal a una señal quasi-rectangular. Podemos variar la Abrupticidad ($s$) y el Offset ($a$).

Observa:
- **Suavidad vs. Ancho de banda**: Cuando la señal es suave ($s$ bajo), varía lentamente y su energía se concentra en frecuencias bajas. Al aumentar la abrupticidad ($s$ alto), las transiciones se vuelven rápidas, lo que obliga a que aparezcan armónicos de orden elevado para construir esos cambios bruscos.
- **Potencia media**: Al hacer la señal más cuadrada (mayor $s$), los coeficientes fundamentales crecen. Esto ocurre porque la señal "se ensancha" en el tiempo, aumentando su área y potencia media[^foot1].
- **Simetría y armónicos pares**: Si el offset es nulo ($a=0$), la señal tiene *simetría de media onda*[^foot2] y sólo contiene armónicos impares. Al añadir asimetría ($a \neq 0$), se rompe esta condición y aparecen los armónicos pares ($k=\pm 2, \pm 4 \dots$).
- **Componente continua** ($c_0$): Este coeficiente representa el valor medio de la señal. Con $a=0$, la señal es simétrica respecto al eje horizontal y su media es nula. Al aumentar el offset, el valor medio deja de ser cero, haciendo que $c_0\neq0$.
:::

[^foot1]: Ver {ref}`subsec:T3:SF:parseval`.
[^foot2]: La **simetría de media onda** ocurre cuando la parte negativa de la señal es un reflejo invertido de la parte positiva, desplazada medio período ($x(t + T_0/2) = -x(t)$). En el dominio espectral, esto causa la cancelación de todos los armónicos pares ($c_2, c_4 \dots$) y la componente continua ($c_0$).

````python
import numpy as np
from bokeh.plotting import figure, show
from bokeh.layouts import column, row
from bokeh.models import ColumnDataSource, CustomJS, Slider, Div
from bokeh.io import output_notebook

# Importamos las funciones auxiliares
from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook(hide_banner=True)

# ==========================================
# 1. PARÁMETROS INICIALES
# ==========================================
VAL_STEEPNESS = 0.5  
VAL_OFFSET = 0.0     # Offset inicial (0 = simétrico)
N_POINTS = 400       
N_HARMONICS = 20     

COLOR_TIME = "blue" 
COLOR_FREQ = "blue" 

t_min, t_max = -2.5, 2.5
t_vals = np.linspace(t_min, t_max, N_POINTS)

# ==========================================
# 2. GENERACIÓN DE DATOS
# ==========================================

# A. Datos Tiempo
# x(t) = tanh( s * (sin(pi*t) + offset) )
def generate_signal(t_arr, s, off):
    if s < 1e-3: s = 1e-3
    # Normalizamos respecto al pico máximo posible para mantener la escala visual
    norm = np.tanh(s * (1.0 + abs(off)))
    return np.tanh(s * (np.sin(np.pi * t_arr) + off)) / norm

y_vals = generate_signal(t_vals, VAL_STEEPNESS, VAL_OFFSET)
source_time = ColumnDataSource(data=dict(t=t_vals, y=y_vals))

# B. Datos Frecuencia
k_vals = np.arange(-N_HARMONICS, N_HARMONICS + 1)
t_period = np.linspace(-1, 1, N_POINTS, endpoint=False) 

def calc_dft(s, off):
    if s < 1e-3: s = 1e-3
    norm = np.tanh(s * (1.0 + abs(off)))
    y_p = np.tanh(s * (np.sin(np.pi * t_period) + off)) / norm
    
    mags = []
    for k in k_vals:
        angle = k * np.pi * t_period
        re = np.mean(y_p * np.cos(angle))
        im = np.mean(y_p * (-np.sin(angle)))
        
        mag = np.sqrt(re**2 + im**2)
        if mag < 1e-4: mag = 0
        mags.append(mag)
        
    return np.array(mags)

mag_vals = calc_dft(VAL_STEEPNESS, VAL_OFFSET)
source_freq = ColumnDataSource(data=dict(k=k_vals, mag=mag_vals, zeros=np.zeros_like(k_vals)))


# ==========================================
# 3. GRÁFICOS
# ==========================================

# --- TIEMPO ---
p_time = figure(width=600, height=300) 
style_math_axes(p_time, x_range=(t_min, t_max), y_range=(-1.3, 1.3), xlabel="t", ylabel=r"$$\tilde{x}(t)$$")
add_math_ticks(p_time, yticks=[-1, 1], ytick_labels=["-1", "1"], tick_len=5)

p_time.line('t', 'y', source=source_time, color=COLOR_TIME, line_width=3)

# --- FRECUENCIA ---
p_freq = figure(width=600, height=300)
style_math_axes(p_freq, x_range=(-N_HARMONICS-1, N_HARMONICS+1), y_range=(0, 0.9), xlabel="k", ylabel=r"$$|c_k|$$")

p_freq.segment(x0='k', y0='zeros', x1='k', y1='mag', source=source_freq, color=COLOR_FREQ, line_width=3)
p_freq.scatter('k', 'mag', source=source_freq, color=COLOR_FREQ, size=8, marker="circle")


# ==========================================
# 4. INTERACTIVIDAD
# ==========================================
s_steep = Slider(start=0.1, end=10.0, value=VAL_STEEPNESS, step=0.1, title=r"Abrupticidad ($$s$$)")
s_offset = Slider(start=0.0, end=1.0, value=VAL_OFFSET, step=0.05, title=r"Offset (Asimetría)")

callback = CustomJS(
    args=dict(source_t=source_time, source_f=source_freq, s_steep=s_steep, s_offset=s_offset),
    code="""
    const s = s_steep.value;
    const off = s_offset.value;
    const PI = Math.PI;
    
    // Normalización dinámica para que la señal no se salga de gráfica
    let norm = Math.tanh(s * (1.0 + Math.abs(off)));
    if (Math.abs(norm) < 1e-9) norm = 1.0;

    // --- 1. TIEMPO ---
    const t = source_t.data['t'];
    const y = source_t.data['y'];

    for (let i = 0; i < t.length; i++) {
        y[i] = Math.tanh(s * (Math.sin(PI * t[i]) + off)) / norm;
    }
    source_t.change.emit();

    // --- 2. FRECUENCIA (DFT) ---
    const N_integ = 200; 
    const k_arr = source_f.data['k'];
    const mag = source_f.data['mag'];
    
    let y_p = new Float32Array(N_integ);
    let t_p = new Float32Array(N_integ);
    
    // Generamos un periodo auxiliar
    for(let i=0; i<N_integ; i++){
        let ti = -1 + (2 * i / N_integ);
        t_p[i] = ti;
        y_p[i] = Math.tanh(s * (Math.sin(PI * ti) + off)) / norm;
    }
    
    for (let i = 0; i < k_arr.length; i++) {
        let k = k_arr[i];
        let sum_re = 0.0;
        let sum_im = 0.0;
        
        for (let j = 0; j < N_integ; j++) {
            let angle = PI * k * t_p[j]; 
            sum_re += y_p[j] * Math.cos(angle);
            sum_im += y_p[j] * (-Math.sin(angle));
        }
        
        let re = sum_re / N_integ;
        let im = sum_im / N_integ;
        let m = Math.sqrt(re*re + im*im);
        
        if (m < 1e-3) m = 0;
        mag[i] = m;
    }
    source_f.change.emit();
""")

s_steep.js_on_change('value', callback)
s_offset.js_on_change('value', callback)

# ==========================================
# 5. LAYOUT
# ==========================================

caption_text = """
<div style="font-family: sans-serif; margin-top: 15px; font-size: 14px; color: #444;">
    <p><b>Interpretación espectral:</b> suavidad/abrupticidad y simetría de la señal.</p>
</div>
"""
caption = Div(text=caption_text)

# Organizamos los sliders en una fila
layout = column(row(s_steep, s_offset), p_time, p_freq, caption, sizing_mode="scale_width")
show(layout)
````

(tip:T3:SF:decaimiento_espectral)=
:::{tip} Regularidad de la señal y decaimiento espectral
Existe una relación matemática directa entre cómo de "suave" (regular) es una señal en el dominio temporal y la rapidez con la que decaen sus armónicos en el dominio frecuencial. 

Como regla general, cada vez que una señal se "suaviza" (matemáticamente, cada vez que se integra), sus coeficientes espectrales se dividen por un factor proporcional a $k$ (concretamente $jk\omega_0$). Veremos la demostración formal de este fenómeno más adelante en la propiedad de integración ({ref}`tab:T3:SF:propiedades`).

En la gráfica interactiva se puede visualizar este efecto. Prestar especial atención a la **línea naranja discontinua**, que representa exclusivamente el **armónico fundamental** ($k=\pm 1$) de la señal:

* **Onda Cuadrada (Discontinua):** Los coeficientes decaen lentamente, del orden de $\mathcal{O}(1/k)$. Como los armónicos superiores aún tienen mucho peso, la señal completa (azul) difiere notablemente de su armónico fundamental (naranja).
* **Onda Triangular (Continua):** Al ser la integral de la onda cuadrada, sus coeficientes decaen más rápido, como $\mathcal{O}(1/k^2)$. La señal completa ya empieza a suavizarse y a acercarse a la forma senoidal.
* **Splines de alto orden (Derivadas continuas):** Al integrar sucesivamente, el decaimiento se vuelve drástico ($\mathcal{O}(1/k^3)$, $\mathcal{O}(1/k^4) \dots$). Esto provoca que los armónicos altos ($k=3, 5 \dots$) se vuelvan diminutos y su aporte sea casi nulo. Como resultado, la señal original queda dominada por completo por el armónico fundamental, volviéndose visualmente indistinguible de una onda senoidal pura.
:::

````python
import numpy as np
from bokeh.plotting import figure, show
from bokeh.layouts import column
from bokeh.models import ColumnDataSource, CustomJS, Select
from bokeh.io import output_notebook
from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook(hide_banner=True)

# ==========================================
# 1. PARÁMETROS Y DATOS EXACTOS
# ==========================================
COLOR_TIME = "blue"
COLOR_FUND = "orange"
COLOR_FREQ = "blue"
N_HARMONICS = 20

# Rango temporal de la gráfica
t_vals = np.linspace(-2.5, 2.5, 2000)
tau = np.mod(t_vals, 2) # Periodo T=2 para definir a trozos

# Vector de tiempo auxiliar de exactamente 1 periodo para la integración
t_1p = np.linspace(0, 2, 2000, endpoint=False)
tau_1p = np.mod(t_1p, 2)

# Armónicos centrados en cero
k_vals = np.arange(-N_HARMONICS, N_HARMONICS + 1)

# Función auxiliar para extraer el armónico fundamental en el tiempo
def get_fundamental(y_period):
    c_1 = np.mean(y_period * np.exp(-1j * np.pi * t_1p))
    return 2 * np.real(c_1 * np.exp(1j * np.pi * t_vals))

# Función auxiliar para extraer TODOS los coeficientes |c_k| reales
def get_all_ck(y_period):
    c_array = np.zeros_like(k_vals, dtype=float)
    for i, k in enumerate(k_vals):
        c = np.mean(y_period * np.exp(-1j * np.pi * k * t_1p))
        # Filtramos ruido numérico para armónicos nulos
        c_array[i] = np.abs(c) if np.abs(c) > 1e-4 else 0.0
    return c_array

# --- DOMINIO DEL TIEMPO ---
# Orden 1: Onda Cuadrada
y_1 = np.where(tau < 1, 1.0, -1.0)
y_1p = np.where(tau_1p < 1, 1.0, -1.0)
f_1 = get_fundamental(y_1p)
c_1 = get_all_ck(y_1p)

# Orden 2: Onda Triangular
y_2 = np.where(tau < 1, tau - 0.5, 1.5 - tau)
y_2 = y_2 / np.max(np.abs(y_2))
y_2p = np.where(tau_1p < 1, tau_1p - 0.5, 1.5 - tau_1p)
y_2p = y_2p / np.max(np.abs(y_2p))
f_2 = get_fundamental(y_2p)
c_2 = get_all_ck(y_2p)

# Orden 3: Spline Parabólico
y_3 = np.where(tau < 1, 0.5*tau**2 - 0.5*tau, -0.5*tau**2 + 1.5*tau - 1.0)
y_3 = y_3 / np.max(np.abs(y_3))
y_3p = np.where(tau_1p < 1, 0.5*tau_1p**2 - 0.5*tau_1p, -0.5*tau_1p**2 + 1.5*tau_1p - 1.0)
y_3p = y_3p / np.max(np.abs(y_3p))
f_3 = get_fundamental(y_3p)
c_3 = get_all_ck(y_3p)

# Orden 4 y 5: Splines Cúbico y Cuártico 
k_series = np.arange(1, 1000, 2)

y_4 = np.sum([(1.0/k**4) * np.cos(k * np.pi * t_vals) for k in k_series], axis=0)
y_4 = y_4 / np.max(np.abs(y_4))
y_4p = np.sum([(1.0/k**4) * np.cos(k * np.pi * t_1p) for k in k_series], axis=0)
y_4p = y_4p / np.max(np.abs(y_4p))
f_4 = get_fundamental(y_4p)
c_4 = get_all_ck(y_4p)

y_5 = np.sum([(1.0/k**5) * np.sin(k * np.pi * t_vals) for k in k_series], axis=0)
y_5 = y_5 / np.max(np.abs(y_5))
y_5p = np.sum([(1.0/k**5) * np.sin(k * np.pi * t_1p) for k in k_series], axis=0)
y_5p = y_5p / np.max(np.abs(y_5p))
f_5 = get_fundamental(y_5p)
c_5 = get_all_ck(y_5p)

all_data = {
    'ord1': {'y': y_1.tolist(), 'fund': f_1.tolist(), 'c': c_1.tolist()},
    'ord2': {'y': y_2.tolist(), 'fund': f_2.tolist(), 'c': c_2.tolist()},
    'ord3': {'y': y_3.tolist(), 'fund': f_3.tolist(), 'c': c_3.tolist()},
    'ord4': {'y': y_4.tolist(), 'fund': f_4.tolist(), 'c': c_4.tolist()},
    'ord5': {'y': y_5.tolist(), 'fund': f_5.tolist(), 'c': c_5.tolist()}
}

# Inicializamos con la onda cuadrada
source_time = ColumnDataSource(data=dict(t=t_vals, y=y_1, fund=f_1))
source_freq = ColumnDataSource(data=dict(k=k_vals, c=c_1, zeros=np.zeros_like(k_vals)))

# ==========================================
# 2. GRÁFICOS
# ==========================================

# --- TIEMPO ---
p_time = figure(width=600, height=300)
style_math_axes(p_time, x_range=(-2.5, 2.5), y_range=(-1.3, 1.8), xlabel="t", ylabel=r"$$\tilde{x}(t)$$")
add_math_ticks(p_time, yticks=[-1, 1], ytick_labels=["-1", "1"], tick_len=5)

p_time.line('t', 'fund', source=source_time, color=COLOR_FUND, line_dash="dashed", line_width=2.5, legend_label="Armónico Fundamental (k=±1)")
p_time.line('t', 'y', source=source_time, color=COLOR_TIME, line_width=3, legend_label="Señal Completa")

p_time.legend.location = "top_right"
p_time.legend.click_policy = "hide"
p_time.legend.background_fill_alpha = 0.8

# --- FRECUENCIA ---
p_freq = figure(width=600, height=300)
# Ajustado rango Y para que 1/2 no quede pegado abajo (máx teórico es ~0.63 para la onda cuadrada)
style_math_axes(p_freq, x_range=(-N_HARMONICS-1, N_HARMONICS+1), y_range=(0, 0.75), xlabel="k", ylabel=r"$$|c_k|$$")

# Marca y etiqueta de 1/2 en la gráfica inferior (tamaño 10pt)
add_math_ticks(p_freq, yticks=[0.5], ytick_labels=[r"$$\frac{1}{2}$$"], tick_len=5, font_size="10pt")

p_freq.segment(x0='k', y0='zeros', x1='k', y1='c', source=source_freq, color=COLOR_FREQ, line_width=3)
p_freq.scatter('k', 'c', source=source_freq, color=COLOR_FREQ, size=8, marker="circle")

# ==========================================
# 3. INTERACTIVIDAD
# ==========================================
opciones = [
    ("ord1", r"Orden 1: Discontinua (Onda Cuadrada, 1/k)"), 
    ("ord2", r"Orden 2: Continua (Onda Triangular, 1/k²)"), 
    ("ord3", r"Orden 3: Derivada 1ª Continua (Spline Parabólico, 1/k³)"),
    ("ord4", r"Orden 4: Derivada 2ª Continua (Spline Cúbico, 1/k⁴)"),
    ("ord5", r"Orden 5: Derivada 3ª Continua (Spline Cuártico, 1/k⁵)")
]

select_signal = Select(title="Nivel de regularidad de la señal:", value="ord1", options=opciones)

callback = CustomJS(
    args=dict(source_t=source_time, source_f=source_freq, all_data=all_data, select=select_signal),
    code="""
    const sig_type = select.value;
    const new_data = all_data[sig_type];
    
    source_t.data['y'] = new_data['y'];
    source_t.data['fund'] = new_data['fund'];
    source_t.change.emit();
    
    source_f.data['c'] = new_data['c'];
    source_f.change.emit();
""")

select_signal.js_on_change('value', callback)

layout = column(select_signal, p_time, p_freq, sizing_mode="scale_width")
show(layout)
````
