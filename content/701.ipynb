# Definición de la transformada Z

En esta página, se describe brevemente la transformada Z, basándose en su comparación con la transformada de Fourier en tiempo discreto y en ventajas respecto a la misma.

#### Introducción

La transformada Z es una alternativa al análisis de la transformada de Fourier en tiempo discreto que se basa en el uso de exponenciales complejas.

Las exponenciales complejas discretas de la forma $z^n$ siendo $z=r \cdot e^{j\Omega}$ (expresada mediante módulo y fase) son autofunciones de los sistemas LTI.

\begin{equation}
 y[n]=x[n]*h[n]= \sum_{k} h[k] z_0^{n-k} = z^n \left(\sum_{k=-\infty}^{\infty} h[k]z_{0}^{-k} \right) = z_{0}^{n} H(z_0),
\end{equation}

siendo $H(z_0)$ la transformada Z de $h[n]$ evaluada en el punto $z_0=|z_0|e^{j\Omega_{0}}$.

Entonces, definimos la **ecuación de análisis de la transformada Z**:

:::{note .simple icon=false} Ecuación de análisis de la transformada Z

\begin{equation}
 X(z) = \sum_{n=-\infty}^{\infty}x[n] z^{-n}.
\end{equation}

:::

- $z_0$ es un **cero** de $X(z)$ si:
```{math}
\lim_{z\to z_0} X(z) = 0.
```
:::
