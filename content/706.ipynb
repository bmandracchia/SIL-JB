# Ejercicios

Los ejercicios de transformada Z, combinados en apartados o de manera aislada, se pueden dividir en:

- Cálculo de **transformada Z**. Aplicando la ecuación de análisis o pares básicos y propiedades.
- Cálculo de **transformada Z inversa**. En la asignatura, se realiza a partir de propiedades y descomposición en fracciones simples, sin necesidad de usar la ecuación de síntesis.
- Determinación de la **región de convergencia (ROC)** a partir de la transformada Z.
- Cálculo de **polos y ceros**, normalmente acompañado del dibujo del diagrama de polos y ceros y debiendo determinar la ROC usando la transformada Z.
- Estudio de **propiedades de sistemas LTI** mediante transformada Z. En concreto, análisis de estabilidad y causalidad, además de estudio de sistemas inversos.
- Análisis de sistemas LTI definidos mediante **ecuaciones en diferencias**. Se puede analizar el sistema a partir de una ecuación dada usando la transformada Z, o bien tener que hallar dicha ecuación a partir de la función de transferencia.

## Ejercicios Resueltos

:::{warning .simple icon=false} Ejercicio 1
:label: ejercicio1:T7:TZ

Considere el sistema LTI causal dado por la siguiente ecuación en diferencias:

$y[n] - \frac{1}{2} y[n-1] = x[n]$.

Obtenga la función de transferencia del sistema, su diagrama de polos y ceros, su región de convergencia, razonando si dicho sistema es estable, y la respuesta al impulso.

:::

:::{tip .simple icon=false class=dropdown} Solución del Ejercicio 1

Aplicando la **propiedad de convolución** de la transformada Z para un sistema LTI obtenemos la función de transferencia a partir de la transformada Z de las señales de entrada y de salida:

\begin{equation*}
 Y(z) = X(Z) \cdot H(z) \Rightarrow H(z) = \frac{Y(z)}{X(z)}. 
\end{equation*}

Para alcanzar una expresión de esta forma, aplicamos las propiedades de **linealidad y desplazamiento temporal** de la transformada Z en la ecuación en diferencias:

\begin{equation*}
 Y(z) - \frac{1}{2} z^{-1} Y(z) = X(z) \Rightarrow \frac{Y(z)}{X(z)} = \boxed{H(z) = \frac{1}{1-\frac{1}{2}z^{-1}}}.
\end{equation*}

Procedemos a analizar la región de convergencia (ROC) del sistema. Como el sistema es causal, sabemos que la ROC se define de la forma $|z| > \alpha$. El caso de la función de transferencia hallada aparece directamente en la tabla de pares básicos de transformada Z, habiendo dos casos posibles:

\begin{equation*}
 X(z) = \frac{1}{1-\alpha z^{-1}} \Rightarrow \text{ROC: } \begin{cases}
   |z| > \alpha,& \text{si } x[n] = \alpha^{n} \cdot u[n], \\
   |z| < \alpha,& \text{si } x[n] = -\alpha^{n} \cdot u[-n-1].
  \end{cases}
\end{equation*}

El primero de los dos casos es el que coincide con la ROC de un sistema causal, planteada anteriormente. Entonces, sustituyendo $\alpha$ por 1/2, obtenemos la ROC del sistema:

\begin{equation*}
 \boxed{\text{ROC: } |z| > \frac{1}{2}}.
\end{equation*}

Rápidamente, a partir de la ROC, podemos determinar la estabilidad del sistema. Como $|z|=1$ está incluida en la ROC, el sistema es **estable**.

Hallamos los polos y los ceros y luego dibujamos el diagrama. Reescribimos $H(z)$ para hallar los polos y los ceros, multiplicando numerador y denominador por $z$ en la expresión obtenida anteriormente.

\begin{equation*}
 H(z) = \frac{z}{z-\frac{1}{2}}.
\end{equation*}

Es sencillo comprobar que esta expresión vale 0 cuando $z=0$ (numerador se anula con denominador no nulo), y su límite tiende a $\infty$ cuando $z \to 1/2$ (denominador se anula con numerador no nulo). Además, si $z \to \infty$, el límite de la función de transferencia es 1, por lo que no hay polos ni ceros en el infinito. Por lo tanto:

\begin{equation*}
 \boxed{\text{Polo en }z=\frac{1}{2} \text{ ; cero en }z=0}.
\end{equation*}

En este caso, es interesante comprobar que hay un cero en el origen, coincidiendo con una condición que debe ocurrir en los sistemas causales. Si no hubiera un cero en el origen, el sistema sería no causal, lo que no es el caso de este ejercicio. Representamos a continuación el diagrama de polos y ceros, incluyendo la ROC.

<img src ="figures/T7/svg/diag_pol_zero_ejres1.svg" width = "400">

Por último, obtenemos la respuesta al impulso del sistema. Para ello, usamos directamente la **tabla de pares básicos de transformada Z**. La información correspondiente ya se puso con anterioridad para analizar la región de convergencia en el sistema causal. Entonces, teniendo en cuenta dicha información:

\begin{equation*}
 \boxed{h[n] = \left( \frac{1}{2} \right)^{n} \cdot u[n]}.
\end{equation*}

:::{warning .simple icon=false} Ejercicio 2
:label: ejercicio2:T7:TZ

Para el mismo sistema LTI causal del {ref}`ejercicio1:T7:TZ`, obtenga todas las respuestas al impulso de todos los sistemas inversos al dado, indicando en cada caso cómo se comporta el sistema inverso respecto de las propiedades de estabilidad y causalidad.

:::

:::{tip .simple icon=false class=dropdown} Solución del Ejercicio 2

Para un sistema LTI invertible, aplicando la propiedad de convolución de la transformada Z:

\begin{equation*}
 h[n] * h_i[n] = \delta[n] \Rightarrow H(z) \cdot H_i(z) = 1. 
\end{equation*}

Cabe anotar que la transformada Z de $\delta[n]$ es 1 (uno de los pares básicos de transformadas Z). Hallamos la función de transferencia de todos los sistemas inversos posibles invirtiendo directamente la primera expresión hallada para $H(z)$.

\begin{equation*}
 H_i(z) = \frac{1}{H(z)} = 1-\frac{1}{2}z^{-1}. 
\end{equation*}

Obtenemos la respuesta al impulso del sistema inverso aplicando la propiedad de **linealidad** y los **pares básicos de transformadas Z**. Para ambos sumandos, solamente existe una transformada Z inversa de acuerdo con las tablas, por lo que solamente hay un sistema inverso. Determinamos su respuesta al impulso:

\begin{equation*}
 \boxed{h_i[n] = \delta[n] - \frac{1}{2}\delta[n-1]}.
\end{equation*}

Hallamos la ROC del sistema inverso a partir de la tabla de propiedades. De acuerdo con la propiedad de linealidad, la ROC de este caso es la intersección de la ROC de cada uno de los dos sumandos. De acuerdo con las tablas de pares básicos de transformadas Z, la ROC del primer sumando ($H_1(z) = 1$) es todo el plano Z, y la del segundo sumando es todo el plano Z excepto el origen. Por lo tanto:

\begin{equation*}
 \text{ROC: } z - \{ 0 \} \equiv |z| > 0.
\end{equation*}

Por lo tanto, como la ROC es de la forma $|z|>\alpha$ y no hay polos en el infinito, el sistema es **causal**. Aunque no se pida explícitamente, podemos hallar los polos y los ceros sabiendo que, en un sistema inverso, los ceros y los polos se corresponden con los polos y los ceros, respectivamente, del sistema original. Entonces, en este caso, el sistema inverso tiene un polo en $z=0$, y un cero en $z=1/2$.

Respecto a la estabilidad, como $|z|=1$ está incluida en la ROC, el sistema es **estable**.

Alternativamente, a partir de la respuesta al impulso obtenida, se puede determinar que el sistema es causal porque su respuesta al impulso solamente se define para valores de $n$ iguales o superiores a 0. También se puede determinar que el sistema es estable porque el sumatorio de todos los términos de $|h_i[n]|$ es menor a $\infty$ ($3/2 < \infty$).

:::{warning .simple icon=false} Ejercicio 3
:label: ejercicio3:T7:TZ

Se define el sistema LTI en tiempo discreto a partir de los siguientes polos y ceros:

Polos: $z_2 = \frac{1}{3}$; $z_3 = -\frac{1}{3}$. Ceros: $z_0 = 1$; $z_1 = \infty$.

Para el sistema que tiene ganancia 1 ($H(0)=9$), obtenga el diagrama de polos y ceros, la función de transferencia, la región de convergencia y la respuesta al impulso asumiendo que el sistema es causal. Analice la estabilidad del sistema.

:::

:::{tip .simple icon=false class=dropdown} Solución del Ejercicio 3
El diagrama de polos y ceros se mostrará al final, junto con la región de convergencia (ROC). Deducimos la función de transferencia a partir de la información dada, recordando la **definición de polos y ceros**:

\begin{equation*}
 \text{Ceros: }\lim_{z \to z_0}H(z) = 0. \quad \text{Polos: }\lim_{z \to z_2}|H(z)| = \infty.
\end{equation*}

\begin{equation*}
 \boxed{H(z) = \frac{z-1}{\left(z-\frac{1}{3} \right) \cdot \left(z+\frac{1}{3} \right)}.}
\end{equation*}

Para hallar la respuesta al impulso, **descomponemos en fracciones simples**:

\begin{equation*}
 H(z) = \frac{A}{z-\frac{1}{3}}+\frac{B}{z+\frac{1}{3}} \rightarrow  \begin{cases}
		A+B=1\\
		\frac{1}{3}A-\frac{1}{3}B=-1
	\end{cases}
\end{equation*}

\begin{equation*}
 \begin{cases}
		A+B=1\\
		A-B=-3
	\end{cases} \Rightarrow 
	\begin{cases}
	A=-1\\
	B=2
    \end{cases}
\end{equation*}

Por tanto,

\begin{equation*}
 H(z) = \frac{-1}{z-\frac{1}{3}}+\frac{2}{z+\frac{1}{3}}=\frac{-z^{-1}}{1-\frac{1}{3}z^{-1}}+\frac{2z^{-1}}{1+\frac{1}{3}z^{-1}}.
\end{equation*}

Para calcular la respuesta al impulso y determinar la ROC, usamos las **propiedades de linealidad y convolución**, además de las tablas de **pares básicos de transformadas Z**, teniendo en cuenta que el sistema es causal (explicación similar en el Ejercicio Resuelto 1).

\begin{equation*}
 h[n] = -\delta[n-1]*\left(\frac{1}{3} \right)^n \cdot u[n] +2\delta[n-1]*\left(-\frac{1}{3} \right)^n \cdot u[n]
\end{equation*}

\begin{equation*}
 \boxed{h[n] = -\left(\frac{1}{3} \right)^{n-1} \cdot u[n-1]+2 \left(-\frac{1}{3} \right)^{n-1} \cdot u[n-1]}
\end{equation*}

\begin{equation*}
 \boxed{h[n] = \frac{1}{3^{n-1}}\left(-1+2(-1)^{n-1} \right) \cdot u[n-1].}
\end{equation*}

Respecto a la ROC, de acuerdo con la propiedad de linealidad, es igual al menos a la intersección de la ROC de ambas fracciones simples, y de acuerdo con la propiedad de convolución, en cada fracción simple la ROC es la intersección de la ROC de la delta desplazada con el término de la potencia definida para tiempos positivos y el 0. Para cada fracción simple, la ROC asociada con $\delta[n-1]$ es todo el plano Z excepto el 0, y la ROC de la fracción es $|z|>\frac{1}{3}$ para las dos fracciones simples. Por lo tanto:

\begin{equation*}
 \text{ROC}_1 = \text{ROC}_2: \{Z-\{0\}\} \cap \left\{|z| > \frac{1}{3} \right\} \Rightarrow \boxed{\text{ROC: } |z| > \frac{1}{3}.}
\end{equation*}

Como $|z| = 1$ está incluida en la ROC, **el sistema es estable**. Mostramos el diagrama de polos y ceros:

Halle la ecuación en diferencias que caracteriza la relación entrada-salida del sistema causal definido en el {ref}`ejercicio3:T7:TZ`.

:::

:::{tip .simple icon=false class=dropdown} Solución del Ejercicio 4

Reescribimos $H(z)$ y aplicamos la propiedad de **convolución** junto con la de **desplazamiento temporal** y la de **linealidad** para hallar el resultado.

\begin{equation*}
 H(z) = \frac{z-1}{z^2-\frac{1}{9}}=\frac{-z^{-2}+z^{-1}}{-\frac{1}{9}z^{-2}+1}
\end{equation*}

\begin{equation*}
 X(z)\cdot H(z) = Y(z) \rightarrow H(z) = \frac{Y(z)}{X(z)}
\end{equation*}

\begin{equation*}
 Y(z) \left(-\frac{1}{9}z^{-2}+1 \right) = X(z) \left(-z^{-2}+z^{-1} \right)
\end{equation*}

\begin{equation*}
 \boxed{-\frac{1}{9}y[n-2]+y[n] = -x[n-2]+x[n-1].}
\end{equation*}

:::{warning .simple icon=false} Ejercicio 5
:label: ejercicio5:T7:TZ

Sin asumir ningún caso concreto de causalidad del sistema definido en el {ref}`ejercicio3:T7:TZ`, obtenga la función de transferencia y dos respuestas al impulso asociadas con el sistema inverso. Analice la estabilidad y causalidad de todos los casos.

:::

:::{tip .simple icon=false class=dropdown} Solución del Ejercicio 5

Un sistema es invertible si (aplicando la propiedad de convolución):

\begin{equation*}
 h[n] * h_i[n] = \delta[n] \Rightarrow H(z) \cdot H_i(z) = 1 \rightarrow H_i(z) = \frac{1}{H(z)},
\end{equation*}

siendo $h_i[n]$ y $H_i(z)$ la respuesta al impulso y la función de transferencia del sistema inverso, respectivamente. Hallamos la función de transferencia del sistema inverso invirtiendo la función de transferencia obtenida anteriormente en el Ejercicio 3 ({ref}`ejercicio3:T7:TZ`):

\begin{equation*}
 H_i(z) = \frac{z^2-\frac{1}{9}}{z-1}=\frac{z-\frac{1}{9}z^{-1}}{1-z^{-1}}=\frac{z}{1-z^{-1}}-\frac{\frac{1}{9}z^{-1}}{1-z^{-1}}.
\end{equation*}

Los **polos y ceros del sistema se corresponden con los ceros y polos del sistema inverso, respectivamente**. Entonces, para cualquiera de las posibles respuestas al impulso, $z=1$ es un polo del sistema inverso, por lo que $|z|=1$ no está incluida en ninguna de las ROC y **en todos los casos el sistema no es estable**. 

De la función de transferencia del sistema inverso extraemos dos respuestas al impulso, una con ROC $|z| > 1$, y otra con ROC $|z|<1$ (sin analizar 0 ni infinito), a partir de las propiedades de **linealidad y convolución**, y las tablas de **pares básicos de transformadas Z**.

\begin{equation*}
 |z|>1 \Rightarrow h[n] = \delta[n+1]*u[n]-\frac{1}{9}\delta[n-1]*u[n]
\end{equation*}

\begin{equation*}
 \boxed{h[n] = u[n+1]-\frac{1}{9}\cdot u[n-1]=\delta[n+1] + \delta[n] +\frac{8}{9}\cdot u[n-1].}
\end{equation*}

\begin{equation*}
 |z|<1 \Rightarrow h[n] = \delta[n+1]*-u[-n-1]-\frac{1}{9}\delta[n-1]*-u[-n-1]
\end{equation*}

\begin{equation*}
 \boxed{h[n] = -u[-n-2]+\frac{1}{9}\cdot u[-n]=\delta[n+1] + \delta[n] -\frac{8}{9}\cdot u[-n].}
\end{equation*}

Respecto a la causalidad, para el primer caso ($|z|>1$), aunque la ROC es de la forma $|z|>\alpha$, como hay un polo en el infinito, el sistema es **no causal**. Para el segundo caso ($|z|<1$), aunque la ROC es de la forma $|z|<\alpha$, como no hay un cero en el origen, el sistema es **no causal**.

Alternativamente, para ambas respuestas al impulso, se puede ver que, al menos, la respuesta al impulso tiene valores no nulos para $n=0$ y $n<0$, por lo que los dos casos son no causales.

:::{warning class=dropdown}Notas adicionales sobre el {ref}`ejercicio5:T7:TZ`

En los resultados para la respuesta al impulso, se puede apreciar que hay dos soluciones equivalentes. El segundo resultado viene de hacer la división de polinomios en lugar de reescribir la expresión mostrada en la resolución. Se procede a realizar el ejercicio mediante la división de polinomios para demostrar dichos resultados.

\begin{equation*}
 H_i(z) = \frac{z^2-\frac{1}{9}}{z-1}= z + 1 + \frac{8/9}{z-1} = z + 1 + \frac{\frac{8}{9}z^{-1}}{1-z^{-1}}.
\end{equation*}

Este resultado daría lugar a dos respuestas al impulso que tendrían en común $\delta[n+1] + \delta[n]$, como se muestra en las soluciones, y un tercer sumando cuyo término en el dominio del tiempo depende de la ROC y se puede encontrar en los pares básicos de transformada Z, aplicando la propiedad de linealidad junto con la de convolución de manera similar a como se mostró anteriormente en la resolución del ejercicio.

También cabe anotar que, partiendo de la expresión de la función de transferencia basada en la suma de dos fracciones, se podría pensar en una tercera (e incluso cuarta) respuesta al impulso siendo una de las ROC (cualquiera de los sumandos) de la forma $|z|>1$, y la otra ROC de la forma $|z|<1$. Sin embargo, por la propiedad de linealidad, como la intersección de ambas ROC es el conjunto vacío, la transformada Z no converge y no son casos que se pudieran considerar realmente.
