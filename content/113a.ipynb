# 1.4.1 Señales exponenciales y senoidales

## Señales exponenciales complejas continuas

:::{important .simple icon=false} Definición general
\begin{equation}
x(t)=C e^{st},\qquad C,s\in\mathbb{C}.
\end{equation}

donde las constantes complejas pueden escribirse como:

\begin{equation}
C=|C|e^{j\theta},
\end{equation}

\begin{equation}
s=\sigma+j\omega_0.
\end{equation}
:::

Sustituyendo estas expresiones en la definición general:

\begin{equation}
x(t)=|C|e^{j\theta}e^{(\sigma+j\omega_0)t}
=|C|e^{\sigma t}e^{j(\omega_0 t+\theta)}.
\end{equation}

Esta expresión muestra que una exponencial compleja combina:

* un término exponencial real $e^{\sigma t}$, que controla el crecimiento o decrecimiento;
* y un término oscilatorio $e^{j(\omega_0 t+\theta)}$, que introduce periodicidad.

---

Dependiendo de los valores de $C$ y $s$, obtenemos distintos tipos de señales.

### Señales exponenciales reales

:::{important .simple icon=false} Señales exponenciales reales
Si
$$
C,s\in\real,
\qquad (s=\sigma),
$$

la señal toma la forma

\begin{equation}
x(t)=C e^{\sigma t},
\qquad C,\sigma\in\real.
\end{equation}
:::

El comportamiento depende del signo de $\sigma$:

* **Creciente** $(\sigma>0)$

  La amplitud aumenta exponencialmente con el tiempo.

  **Ejemplos:** crecimiento poblacional, reacciones en cadena.

* **Decreciente** $(\sigma<0)$

  La amplitud disminuye exponencialmente.

  **Ejemplos:** descarga de un condensador en un circuito RC, sistemas mecánicos amortiguados.

* **Constante** $(\sigma=0)$

  La señal no cambia con el tiempo:
  \begin{equation}
  x(t)=C.
  \end{equation}

### Señales exponenciales imaginarias o senoidales

:::{important .simple icon=false} Señales exponenciales imaginarias o senoidales
Si el exponente es puramente imaginario,

$$
s=j\omega_0,
$$

la señal se escribe como

\begin{equation}
x(t)=C e^{j\omega_0 t},
\qquad C\in\mathbb{C},\ \omega_0\in\real.
\end{equation}
:::

Usando la forma polar de $C$:

\begin{equation}
x(t)=|C|e^{j\theta}e^{j\omega_0 t}
=|C|e^{j(\omega_0 t+\theta)}.
\end{equation}

El caso más simple es:

\begin{equation}
x(t)=e^{j\omega_0 t}.
\end{equation}

Gracias a la fórmula de Euler,

\begin{equation}
e^{j\phi}=\cos(\phi)+j\sin(\phi),
\end{equation}

podemos escribir:

\begin{equation}
e^{j\omega_0 t}
=\cos(\omega_0 t)+j\sin(\omega_0 t).
\end{equation}

Por tanto, una exponencial compleja representa una oscilación formada por:

* una parte real cosenoidal;
* y una parte imaginaria senoidal.

:::{warning} Atención
Aunque se llamen *exponenciales imaginarias*, estas señales **no producen números imaginarios puros**.

Por ejemplo,

\begin{equation}
e^{j\omega_0 t}
=\cos(\omega_0 t)+j\sin(\omega_0 t),
\end{equation}

tiene simultáneamente:

* una parte real: $\cos(\omega_0 t)$;
* y una parte imaginaria: $\sin(\omega_0 t)$.

La señal va recorriendo distintos puntos del plano complejo a medida que cambia el tiempo.
:::

#### Representaciones de la señal $ x(t)=e^{j\omega_0 t} $

Una señal exponencial compleja puede representarse de distintas formas, cada una de ellas resaltando una propiedad diferente de la señal. En esta sección se muestran tres representaciones complementarias:

- **Representación tridimensional**, que combina el tiempo y el plano complejo para visualizar simultáneamente la rotación del fasor y su evolución temporal.
- **Representación temporal**, que muestra la evolución de sus partes real e imaginaria en función del tiempo.
- **Representación en el plano complejo (fasor)**, donde la señal se interpreta como un vector de módulo constante que gira con velocidad angular $ \omega_0 $.

Estas tres perspectivas son equivalentes y ayudan a comprender la relación entre la expresión matemática de la señal y su interpretación geométrica.

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ============================================================
# Parámetros
# ============================================================

t = np.linspace(0, 4*np.pi, 600)
omegas = np.linspace(1, 10.0, 11)

# ============================================================
# Función
# ============================================================

def exp_imaginaria(t, omega):
    x = np.exp(1j * omega * t)

    re = np.real(x)
    im = np.imag(x)

    A = np.abs(x)

    # Fase envuelta entre -pi y pi
    phi = np.angle(x)

    return re, im, A, phi

# ============================================================
# Figura vertical
# ============================================================

fig = make_subplots(
    rows=3,
    cols=1,
    vertical_spacing=0.12,
    specs=[
        [{"type": "scene"}],
        [{"type": "xy"}],
        [{"type": "xy"}],
    ],
    subplot_titles=[
        "Representación 3D",
        "Parte real e imaginaria",
        "Amplitud y fase"
    ]
)

# ============================================================
# Datos iniciales
# ============================================================

omega0 = omegas[0]
re, im, A, phi = exp_imaginaria(t, omega0)

# ============================================================
# Gráfica 3D
# ============================================================

fig.add_trace(
    go.Scatter3d(
        x=t,
        y=re,
        z=im,
        mode="lines",
        line=dict(
            width=6,
            # color=t,
            # colorscale="Viridis",
        ),
    ),
    row=1,
    col=1
)

# ============================================================
# Re e Im
# ============================================================

fig.add_trace(
    go.Scatter(
        x=t,
        y=re,
        mode="lines",
        name="Re{x(t)}",
    ),
    row=2,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=t,
        y=im,
        mode="lines",
        name="Im{x(t)}",
    ),
    row=2,
    col=1
)

# ============================================================
# Amplitud y fase
# ============================================================

fig.add_trace(
    go.Scatter(
        x=t,
        y=A,
        mode="lines",
        name="A(t)",
    ),
    row=3,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=t,
        y=phi,
        mode="lines",
        name="φ(t)",
    ),
    row=3,
    col=1
)

# ============================================================
# Frames para animación
# ============================================================

frames = []

for omega in omegas:

    re, im, A, phi = exp_imaginaria(t, omega)

    frames.append(
        go.Frame(
            data=[
                go.Scatter3d(x=t, y=re, z=im),
                go.Scatter(x=t, y=re),
                go.Scatter(x=t, y=im),
                go.Scatter(x=t, y=A),
                go.Scatter(x=t, y=phi),
            ],
            name=f"{omega:.2f}"
        )
    )

fig.frames = frames

# ============================================================
# Slider
# ============================================================

sliders = [
    dict(
        active=0,
        currentvalue=dict(prefix="ω₀ = "),
        steps=[
            dict(
                label=f"{omega:.2f}",
                method="animate",
                args=[
                    [f"{omega:.2f}"],
                    dict(
                        mode="immediate",
                        frame=dict(duration=0, redraw=True),
                        transition=dict(duration=0)
                    )
                ],
            )
            for omega in omegas
        ],
    )
]

# ============================================================
# Layout
# ============================================================
bgcolor = "whitesmoke"

fig.update_layout(
    # title=r"Representaciones de la señal $x(t)=e^{j\omega_0 t}$",
    sliders=sliders,
    width=600,
    height=1200,
    plot_bgcolor=bgcolor,
)

# ============================================================
# Ejes 3D
# ============================================================

fig.update_scenes(
    xaxis_title="t",
    yaxis_title="Re{x(t)}",
    zaxis_title="Im{x(t)}",xaxis=dict(
        backgroundcolor=bgcolor,
    ),

    yaxis=dict(
        backgroundcolor=bgcolor,
    ),

    zaxis=dict(
        backgroundcolor=bgcolor,
    ),
)


# ============================================================
# Ejes 2D
# ============================================================

fig.update_xaxes(title_text="t", row=2, col=1)
fig.update_yaxes(title_text="Valor", row=2, col=1)

fig.update_xaxes(title_text="t", row=3, col=1)

fig.update_yaxes(
    title_text="A(t), φ(t)",
    range=[-np.pi, np.pi],
    tickvals=[-np.pi, -np.pi/2, 0, np.pi/2, np.pi],
    ticktext=["−π", "−π/2", "0", "π/2", "π"],
    row=3,
    col=1
)

fig.show()
````

Son **señales periódicas**, ya que cumplen la ecuación  [](#periodica_continua):
```{math}
x(t)=x(t+T),\ \forall t,
```
```{math}
\begin{equation}
  |C|e^{j\theta}e^{j\omega_0 t}=|C|e^{j\theta}e^{j\omega_0 (t+T)}=|C|e^{j\theta}e^{j\omega_0 t}e^{j\omega_0 T}
\end{equation}
```
\begin{equation*}
 \Downarrow 
\end{equation*} 
\begin{equation}\label{condicion_exp_periodica}
  e^{j\omega_o T}=1.
\end{equation}
Si $\omega_0=0$, la señal es constante, $x(t)=C$, y es periódica, para cualquier período, $T$.

Si $\omega_0\neq 0$, para que se cumpla la condición dada por la ecuación [](#condicion_exp_periodica), $\omega_0 T$ debe ser múltiplo de $2\pi$, es decir:
\begin{equation}
  \omega_0 T=2\pi m,\ m\in\Z.
\end{equation}
Según la definición vista de período fundamental (valor mínimo del período), éste debe cumplir:
\begin{equation}
  \omega_0 T=\pm 2\pi,
\end{equation}
por lo que el período fundamental de una exponencial compleja continua es:
\begin{equation}\label{período_frecuencia}
  T_0=\frac{2\pi}{|\omega_0|}.
\end{equation}
$\omega_0$ es la frecuencia fundamental, en rad/s, de la exponencial compleja periódica, y está relacionada con la frecuencia fundamental en Hertzios:
\begin{equation}
  \omega_0=2\pi f_0.
\end{equation}
Las exponenciales $e^{j\omega_0 t}$ y $e^{-j\omega_0 t}$ tienen el mismo período fundamental, $T_0$.

Una señal muy relacionada con la exponencial compleja periódica son las señales senoidales (seno y coseno):
\begin{equation}
  x(t)=A\cos(\omega_0 t +\theta).
\end{equation}

En la ecuación [](#período_frecuencia) se observa que el período fundamental y la frecuencia fundamental son inversamente proporcionales. Como se ve en la siguiente gráfica, si se aumenta la frecuencia fundamental, aumenta la velocidad de oscilación y disminuye el período fundamental, y viceversa. Además, este incremento en la velocidad de oscilación con la frecuencia fundamental es monótono, es decir, siempre que se aumente la frecuencia fundamental de la señal exponencial compleja, aumenta la velocidad de oscilación.

A partir de la relación de Euler:
\begin{equation} \label{ecuacion_euler}
  e^{j\omega_0 t}=\cos(\omega_0 t)+j\sin(\omega_0 t),
\end{equation}
las señales senoidales también pueden ponerse en forma de exponencial compleja periódica:

  \begin{equation}
    A\cos(\omega_0 t+\theta)=\frac{A}{2}e^{j\theta}e^{j\omega_0 t}+\frac{A}{2}e^{-j\theta}e^{-j\omega_0 t}=A\Re\{e^{j(\omega_0 t+\theta)}\},
  \end{equation}
  \begin{equation}
    A\sin(\omega_0 t+\theta)=\frac{A}{2j}e^{j\theta}e^{j\omega_0 t}-\frac{A}{2j}e^{-j\theta}e^{-j\omega_0 t}=A\Im\{e^{j(\omega_0 t+\theta)}\}.
  \end{equation}


Este tipo de señales aparecen en los sistemas físicos con conservación de energía, como por ejemplo, en la respuesta natural de un circuito LC, en el movimiento armónico simple o en la presión acústica de una nota musical.


Las señales periódicas en general, y éstas en particular, son ejemplos de señales de potencia (hacemos el cálculo para $x(t)=e^{j\omega_0 t}$):
\begin{equation}
  E_0=\int_0^{T_0}|e^{j\omega_o t}|^2 dt=\int_0^{T_0} 1dt=T_0.
\end{equation}
Como hay un número infinito de períodos, al integrar a todo el tiempo, obtenemos una energía total:
\begin{equation}
  E_{\infty}=\int_{-\infty}^{\infty}|e^{j\omega_o t}|^2 dt=\infty.
\end{equation}
La potencia media la podemos calcular con la expresión para señales periódicas:
\begin{equation}
  P_\infty=\frac{1}{T_0}\int_0^{T_0}|e^{j\omega_o t}|^2 dt=\frac{E_{T_0}}{T_0}=1,
\end{equation}
o bien con la expresión general:
\begin{equation}
  P_\infty=\lim_{T\to\infty}\frac{1}{2T}\int_{-T}^{T}|e^{j\omega_o t}|^2 dt=\lim_{T\to\infty}\frac{2T}{2T}=1.
\end{equation}


Las señales periódicas exponenciales complejas son fundamentales para el análisis de señales y sistemas, pues se pueden emplear como base para muchas otras señales, como veremos en el tema 4.

Para ello es útil considerar **conjuntos de exponenciales complejas relacionadas armónicamente**:
\begin{equation}
  \phi_k(t)=e^{j\omega_k t}.
\end{equation}
Todas las exponenciales complejas relacionadas armónicamente tienen la propiedad de ser periódicas con un período común $T_0$[^1]:
[^1]:$T_0$ no es ahora el período fundamental de cada una, sino el período común de todas ellas.
\begin{equation}
  x(t)=x(t+T_0)\ \Rightarrow\ e^{j\omega_k T_0}=1.
\end{equation}
\begin{equation}
  \omega_k T_0=2\pi k,\ k\in\Z.
\end{equation}
Si definimos $\omega_0$ como[^2]:
[^2]: $\omega_0$ tampoco es ahora la frecuencia fundamental de cada una, que será $\omega_k$.
\begin{equation}
  \omega_0=\frac{2\pi}{T_0},
\end{equation}
la frecuencia fundamental de cada una de las exponenciales complejas relacionadas armónicamente es:
\begin{equation}
  \omega_k=k\omega_0.
\end{equation}
Sus frecuencias fundamentales, $\omega_k$, son múltiplos enteros de una sola frecuencia, $\omega_0>0$.
\begin{equation}
  \phi_k(t)=e^{jk\omega_0 t}=e^{jk\frac{2\pi}{T_0} t}.
\end{equation}

El período fundamental de cada exponencial compleja relacionada armónicamente será:
\begin{equation}
  \frac{2\pi}{|\omega_k|}=\frac{2\pi}{|k|\omega_0}=\frac{T_0}{|k|},\ k\in\Z,
\end{equation}
por lo que, como ya hemos dicho, $\phi_k(t)$ es también periódica con período $T_0$ (múltiplo del período fundamental).

Además, dado que $k\in\Z$, hay infinitas exponenciales complejas relacionadas armónicamente con un cierto período $T_0$.

### Señales exponenciales complejas generales

:::{important .simple icon=false} Señales exponenciales complejas generales
Tienen la forma más general:
\begin{equation}
x(t)=C e^{st},\quad C,s\in\mathbb{C},
\end{equation}

donde $s$ está dado en forma rectangular y $C$ en forma polar:

  \begin{equation}
    C=|C|e^{j\theta},
  \end{equation}
  \begin{equation}
    s=\sigma+j\omega_0.
  \end{equation}
:::

Podemos interpretarlas a partir de los dos casos anteriores:
\begin{equation}
  x(t)=Ce^{st}=|C|e^{j\theta}e^{(\sigma+j\omega_0)t}=|C|e^{\sigma t}e^{j(\omega_o t+\theta)},
\end{equation}
y a partir de la ecuación de Euler  [](#ecuacion_euler):
\begin{equation}
  x(t)=|C|e^{\sigma t}\cos(\omega_o t+\theta)+j|C|e^{\sigma t}\sin(\omega_o t+\theta).
\end{equation}
Corresponden a señales oscilantes crecientes o decrecientes, con envolvente:
\begin{equation}
  \pm|C|e^{\sigma t}.
\end{equation}

Según sea el signo de $\sigma$, tendremos:

  - $\sigma=0$: exponencial compleja periódica (partes real e imaginaria senoidales periódicas).
  - $\sigma>0$: exponencial compleja periódica por exponencial creciente[^3] (partes real e imaginaria senoidal por exponencial creciente).
  - $\sigma<0$: exponencial compleja periódica por exponencial decreciente (partes real e imaginaria senoidal por exponencial decreciente).

[^3]: El resultado ya no es periódico

#### Representaciones de la señal exponencial compleja general

La señal exponencial compleja general

$$
x(t)=e^{-\alpha t}e^{j\omega_0 t}
$$

combina una oscilación de frecuencia angular  $ \omega_0 $ con una envolvente exponencial determinada por el parámetro $ \alpha $. Dependiendo del signo de $ \alpha $, la amplitud de la señal puede decrecer o crecer con el tiempo.

Al igual que en el caso de la exponencial compleja pura, esta señal puede representarse de distintas formas, cada una de ellas destacando un aspecto diferente de su comportamiento:

- **Representación tridimensional**, que combina el tiempo y el plano complejo para visualizar simultáneamente la rotación del fasor y la evolución de su módulo.
- **Representación temporal**, que muestra la evolución de las partes real e imaginaria y cómo varía su amplitud con el tiempo.
- **Representación en el plano complejo**, donde el fasor describe una espiral cuyo módulo cambia exponencialmente mientras continúa girando con velocidad angular $ \omega_0 $.

Estas tres representaciones permiten comprender la relación entre la expresión analítica de la señal y su interpretación geométrica, mostrando cómo la envolvente exponencial modifica el movimiento circular uniforme de la exponencial compleja pura.


````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ============================================================
# Parámetros
# ============================================================

t = np.linspace(0, 4*np.pi, 700)

omegas = np.linspace(1, 10, 10)
alphas = np.linspace(0.0, 1.0, 6)

bgcolor = "whitesmoke"

# ============================================================
# Señal
# x(t) = e^{-alpha t} e^{j omega t}
# ============================================================

def exp_general(t, omega, alpha):
    x = np.exp(-alpha*t) * np.exp(1j*omega*t)

    re = np.real(x)
    im = np.imag(x)
    A = np.abs(x)
    phi = np.angle(x)

    return re, im, A, phi

# ============================================================
# Figura
# ============================================================

fig = make_subplots(
    rows=3,
    cols=1,
    vertical_spacing=0.12,
    specs=[
        [{"type": "scene"}],
        [{"type": "xy"}],
        [{"type": "xy"}],
    ],
    subplot_titles=[
        "Representación 3D",
        "Parte real e imaginaria",
        "Amplitud y fase"
    ]
)

# ============================================================
# Datos iniciales
# ============================================================

omega0 = omegas[0]
alpha0 = alphas[1]

re, im, A, phi = exp_general(t, omega0, alpha0)

# 3D
fig.add_trace(
    go.Scatter3d(
        x=t,
        y=re,
        z=im,
        mode="lines",
        name="x(t)",
        line=dict(width=6),
    ),
    row=1,
    col=1
)

# Re e Im
fig.add_trace(go.Scatter(x=t, y=re, mode="lines", name="Re{x(t)}"), row=2, col=1)
fig.add_trace(go.Scatter(x=t, y=im, mode="lines", name="Im{x(t)}"), row=2, col=1)

# Envolvente en Re/Im
fig.add_trace(
    go.Scatter(x=t, y=A, mode="lines", name="+A(t)", line=dict(dash="dash")),
    row=2,
    col=1
)

fig.add_trace(
    go.Scatter(x=t, y=-A, mode="lines", name="-A(t)", line=dict(dash="dash")),
    row=2,
    col=1
)

# Amplitud y fase
fig.add_trace(go.Scatter(x=t, y=A, mode="lines", name="A(t)"), row=3, col=1)
fig.add_trace(go.Scatter(x=t, y=phi, mode="lines", name="φ(t)"), row=3, col=1)

# ============================================================
# Frames: una combinación por frame
# ============================================================

frames = []
slider_steps = []

for alpha in alphas:
    for omega in omegas:

        re, im, A, phi = exp_general(t, omega, alpha)

        frame_name = f"ω={omega:.1f}, α={alpha:.2f}"

        frames.append(
            go.Frame(
                data=[
                    go.Scatter3d(x=t, y=re, z=im),
                    go.Scatter(x=t, y=re),
                    go.Scatter(x=t, y=im),
                    go.Scatter(x=t, y=A),
                    go.Scatter(x=t, y=-A),
                    go.Scatter(x=t, y=A),
                    go.Scatter(x=t, y=phi),
                ],
                name=frame_name,
            )
        )

        slider_steps.append(
            dict(
                label=frame_name,
                method="animate",
                args=[
                    [frame_name],
                    dict(
                        mode="immediate",
                        frame=dict(duration=0, redraw=True),
                        transition=dict(duration=0),
                    ),
                ],
            )
        )

fig.frames = frames

# ============================================================
# Slider único
# ============================================================

sliders = [
    dict(
        active=10,
        currentvalue=dict(prefix="Parámetros: "),
        steps=slider_steps,
    )
]

# ============================================================
# Layout
# ============================================================

fig.update_layout(
    # title=r"Señal exponencial compleja general: $x(t)=e^{-\alpha t}e^{j\omega_0 t}$",
    width=600,
    height=1200,
    sliders=sliders,
    showlegend=True,
    plot_bgcolor=bgcolor,
    # paper_bgcolor=bgcolor,
)

# ============================================================
# Ejes 3D
# ============================================================

fig.update_scenes(
    # bgcolor=bgcolor,
    xaxis=dict(
        title="t",
        backgroundcolor=bgcolor,
    ),
    yaxis=dict(
        title="Re{x(t)}",
        range=[-1.1, 1.1],
        backgroundcolor=bgcolor,
    ),
    zaxis=dict(
        title="Im{x(t)}",
        range=[-1.1, 1.1],
        backgroundcolor=bgcolor,
    ),
)

# ============================================================
# Ejes 2D
# ============================================================

fig.update_xaxes(
    title_text="t",
    range=[0, 4*np.pi],
    tickvals=[0, np.pi, 2*np.pi, 3*np.pi, 4*np.pi],
    ticktext=["0", "π", "2π", "3π", "4π"],
    row=2,
    col=1
)

fig.update_yaxes(
    title_text="Valor",
    range=[-1.1, 1.1],
    tickvals=[-1, 0, 1],
    row=2,
    col=1
)

fig.update_xaxes(
    title_text="t",
    range=[0, 4*np.pi],
    tickvals=[0, np.pi, 2*np.pi, 3*np.pi, 4*np.pi],
    ticktext=["0", "π", "2π", "3π", "4π"],
    row=3,
    col=1
)

fig.update_yaxes(
    title_text="A(t), φ(t)",
    range=[-np.pi, np.pi],
    tickvals=[-np.pi, -np.pi/2, 0, np.pi/2, np.pi],
    ticktext=["−π", "−π/2", "0", "π/2", "π"],
    row=3,
    col=1
)

fig.show()
````


## Señal exponenciales complejas discretas
:::{important .simple icon=false} Señal exponencial compleja discreta:
\begin{equation}
x[n]=C z^n,\quad C, z\in\mathbb{C}.
\end{equation}

\begin{equation}
  C=|C|e^{j\theta},
\end{equation}
\begin{equation}
  z=|z|e^{j\Omega_0}.
\end{equation}
:::

Para que sea una expresión más análoga al caso continuo también se puede poner de la siguiente forma, aunque es menos habitual:
\begin{equation}
x[n]=C e^{\beta n},\quad C, \beta\in\mathbb{C},
\end{equation}
donde
\begin{equation}
z=e^{\beta}.
\end{equation}

Dependiendo de las constantes $C$ y $z$, podemos tener señales con distintas características:

### Señales exponenciales reales
:::{important .simple icon=false} Señales exponenciales reales
$$C, z=r\in\real$$
:::

Según el valor de $r$:

  - $|r|>1$: crece exponencialmente.
  - $|r|<1$: decrece exponencialmente.
  - $r>0$: valores del mismo signo.
  - $r<0$: valores con signo alterno.
  - $r=1$: valor constante.
  - $r=-1$: valor alterno $\pm C$.


```{figure} figures/T1/2_3_fig8 
```


### Señales senoidales
:::{important .simple icon=false} Señales senoidales
 $$|z|=1$$
\begin{equation}
  x[n]=Ce^{j\Omega_0 n},\quad C\in\mathbb{C},\ \Omega_0\in\real.
\end{equation}
:::

Podemos expresarla de la siguiente forma:
\begin{equation}
  x[n]=|C|e^{j\theta}e^{j\Omega_0 n}=|C|e^{j(\Omega_0 n+\theta)}.
\end{equation}
El caso más sencillo corresponde a $C=1$ (ecuación de Euler [](#ecuacion_euler)):
\begin{equation}
  e^{j\Omega_0 n}=\cos(\Omega_0 n)+j\sin(\Omega_0 n).
\end{equation}
al igual que en el caso continuo, está muy relacionada con las señales senoidales:
\begin{equation}
  x[n]=A\cos(\Omega_0 n+\theta)=\frac{A}{2}e^{j\theta}e^{j\Omega_0 n}+\frac{A}{2}e^{-j\theta}e^{-j\Omega_0 n}.
\end{equation}
De forma equivalente al caso continuo, son **señales de potencia** ($E_\infty$ infinita, $P_\infty$ finita), pero en este caso no son necesariamente periódicas, como veremos en la sección [](#periodicidad_exp_discretas).

```{figure} figures/T1/2_3_fig9 
---
width: 70.0%
---
```

### Señales exponenciales complejas generales

:::{important .simple icon=false} Señales exponenciales complejas generales
Tienen la forma más general:
\begin{equation}
  x[n]=Cz^n,\quad C, z\in\mathbb{C},
\end{equation}
donde $C$ y $z$ están dados en forma polar:

  \begin{equation}
    C=|C|e^{j\theta},
  \end{equation}
  \begin{equation}
    z=|z|e^{j\Omega_0}.
  \end{equation}
:::

Podemos interpretarlas a partir de los dos casos anteriores, expresando:
\begin{equation}
  x[n]=Cz^n=|C||z|^n e^{j(\Omega_0 n+\theta)}=|C||z|^n \cos(\Omega_0 n+\theta)+j|C||z|^n \sin(\Omega_0 n+\theta).
\end{equation}
Según sea $|z|$ tenemos distintos casos:

  - $|z|=1$: señal senoidal (partes real e imaginaria senoidales).
  - $|z|>1$: señal senoidal por exponencial creciente.
  - $|z|<1$: señal senoidal por exponencial decreciente.


```{figure} figures/T1/2_3_fig10 
```

## Propiedades de periodicidad de las exponenciales discretas
Para la señal periódica exponencial compleja continua hemos visto que tiene las siguientes propiedades:

- $e^{j\omega_0 t}$ es periódica, $\forall$ $\omega_0$.
- Exponenciales complejas distintas para distintos valores de $\omega_0$. Cuanto mayor sea $\omega_0$, mayor será la velocidad de oscilación de la señal.
- Existen infinitas exponenciales complejas relacionadas armónicamente con un cierto período dado $T_0$.

Vamos a ver que estas propiedades son distintas para el caso de las exponenciales complejas senoidales discretas, lo que hará que haya que tener cierto cuidado al manejarlas.

Estudiaremos, sin pérdida de generalidad, las propiedades de $x[n]=e^{j\Omega_0 n}$.

### Periodicidad

Aunque una senoidal discreta oscile, y tenga envolvente periódica, no tiene por qué ser necesariamente periódica, pues dependerá de donde estén colocadas las muestras. Sabemos que para que sea periódica se debe cumplir:
```{math}
x[n]=x[n+N],
```
\begin{equation}
  e^{j\Omega_0 n}=e^{j\Omega_0(n+N)}\ \Rightarrow\ e^{j\Omega_0 N}=1,
\end{equation}
que se cumple sólo en el caso de que se cumpla la siguiente relación:
\begin{equation}
  \Omega_0 N=2\pi m,\ m\in\Z.
\end{equation}
Por lo tanto, **para que una señal senoidal discreta sea periódica**, debe cumplirse la siguiente condición:
\begin{equation}
  \frac{\Omega_0}{2\pi}=\frac{m}{N},
\end{equation}
o lo que es lo mismo, que esta expresión sea racional.

En este caso, podemos obtener el **período fundamental**, $N_0$, que será el mínimo valor del período:
\begin{equation}
  N_0=\min\left\{\frac{2\pi}{\Omega_0}m\right\},\ m\in\Z.
\end{equation}
La **frecuencia fundamental** ahora no será $\Omega_0$, como en el caso continuo, sino:
\begin{equation}
  \frac{\Omega_0}{m}=\frac{2\pi}{N_0}.
\end{equation}

Veamos algunos ejemplos:

**Ejemplo 1**: $x_1[n]=\cos(2\pi n/12),\quad \Omega_0=2\pi/12=\pi/6$.

$x_1[n]$ se puede ver como muestras de $x_1(t)=\cos(2\pi t/12)$ cada $t=1 s.$ Esta señal tiene frecuencia fundamental $\omega_0=\pi/6$ rad/s y período fundamental $T_0=12 s.$

El período fundamental será:
```{math}
N_0=\min\left\{\frac{2\pi}{\Omega_0}m\right\}=\min\left\{\frac{2\pi}{\pi/6}m\right\}=\min\left\{12m\right\},\ m\in\Z\ \overset{m=1}{\Longrightarrow} N_0=12.
```
La señal $x_1[n]$ se repite cada 12 puntos, o lo que es lo mismo, cada período de la envolvente.

```{figure} figures/T1/2_3_fig12 
---
width: 80.0%
---
```



**Ejemplo 2**: $x_2[n]=\cos(8\pi n/31),\quad \Omega_0=8\pi/31$.

$x_2[n]$ se puede ver como muestras de $x_2(t)=\cos(8\pi t/31)$ cada $t=1 s.$ Esta señal tiene frecuencia fundamental $\omega_0=8\pi/31$ rad/s y período fundamental $T_0=31/4 s.$

El período fundamental será:
```{math}
N_0=\min\left\{\frac{2\pi}{\Omega_0}m\right\}=\min\left\{\frac{2\pi}{8\pi/31}m\right\}=\min\left\{\frac{31}{4}m\right\},\ m\in\Z\ \overset{m=4}{\Longrightarrow} N_0=31.
```
La señal $x_2[n]$ se repite cada 31 puntos, o lo que es lo mismo, cada cuatro períodos de la envolvente.


```{figure} figures/T1/2_3_fig13 
---
width: 90.0%
---
```



**Ejemplo 3**: $x_3[n]=\cos(2 n/5),\quad \Omega_0=2/5$.

$x_3[n]$ se puede ver como muestras de $x_3(t)=\cos(2 t/5)$ cada $t=1 s.$ Esta señal tiene frecuencia fundamental $\omega_0=2/5$ rad/s y período fundamental $T_0=5\pi s.$

Tratamos de calcular el período fundamental de la señal discreta:
```{math}
N_0=\min\left\{\frac{2\pi}{\Omega_0}m\right\}=\min\left\{\frac{2\pi}{2/5}m\right\}=\min\left\{5\pi m\right\},\ m\in\Z\ \Rightarrow\ \text{No periódica}.
```

Aunque la envolvente, $x_3(t)$, es una señal periódica, la señal discreta, $x_3[n]$, no lo es.




**Ejemplo 4**: Consideremos ahora la suma de dos exponenciales complejas periódicas:

$x[n]=e^{j\frac{2\pi}{3}n}+e^{j\frac{3\pi}{4}n}.$

El primer término tiene período fundamental $N_1=3$, mientras que para el segundo término $N_2=8$. La suma de ambos es periódica y tendrá como **período el mínimo común múltiplo (mcm) de ambos**.

\begin{equation}
  N_0=\text{mcm}\{N_1,N_2\}=24.
\end{equation}

````python
import numpy as np

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, CustomJS, Select, Div, Slider
from bokeh.layouts import column, row

from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook(verbose=False, hide_banner=True)

# ==============================================================================
# 1. EJE DISCRETO
# ==============================================================================

n_min, n_max = -18, 18
n = np.arange(n_min, n_max + 1)

initial_case = "Periódica: Ω₀ = π/4"
N0 = 8

# ==============================================================================
# 2. CASOS
# ==============================================================================

cases = {
    "Periódica: Ω₀ = π/4": {
        "omega": np.pi / 4,
        "label": "π/4",
        "periodic": True,
        "period": 8,
    },

    "Periódica: Ω₀ = 2π/5": {
        "omega": 2 * np.pi / 5,
        "label": "2π/5",
        "periodic": True,
        "period": 5,
    },

    "No periódica: Ω₀ = π/√2": {
        "omega": np.pi / np.sqrt(2),
        "label": "π/√2",
        "periodic": False,
        "period": None,
    },

    "No periódica: Ω₀ = 2/5": {
        "omega": 2 / 5,
        "label": "2/5",
        "periodic": False,
        "period": None,
    },
}

case0 = cases[initial_case]
omega0 = case0["omega"]

x0 = np.cos(omega0 * n)
x_shift0 = np.cos(omega0 * (n + N0))

source = ColumnDataSource(
    data=dict(
        n=n,
        zero=np.zeros_like(n, dtype=float),
        x=x0,
        x_shift=x_shift0,
    )
)

# ==============================================================================
# 3. FIGURA
# ==============================================================================

p = figure(
    height=470,
    width=650,
    title="Periodicidad de las sinusoides discretas",
    tools="pan,wheel_zoom,box_zoom,reset"
)

# Señal original
p.segment(
    "n",
    "zero",
    "n",
    "x",
    source=source,
    line_width=2,
    color="navy",
    alpha=0.35
)

p.scatter(
    "n",
    "x",
    source=source,
    size=9,
    color="navy",
    legend_label="x[n]"
)

# Señal desplazada
p.segment(
    "n",
    "zero",
    "n",
    "x_shift",
    source=source,
    line_width=3,
    color="firebrick",
    alpha=0.45
)

p.scatter(
    "n",
    "x_shift",
    source=source,
    size=11,
    marker="diamond",
    color="firebrick",
    fill_alpha=0.25,
    line_width=2,
    legend_label="x[n+N]"
)

p.line(
    "n",
    "zero",
    source=source,
    color="black",
    alpha=0.35
)

p.legend.location = "top_right"
p.legend.click_policy = "hide"
p.grid.grid_line_alpha = 0.25

style_math_axes(
    p,
    x_range=(n_min - 1, n_max + 1),
    y_range=(-1.35, 1.35),
    xlabel="n",
    ylabel="amplitud"
)

add_math_ticks(
    p,
    yticks=[-1, 1],
    ytick_labels=["-1", "1"],
    tick_len=5
)

# ==============================================================================
# 4. CONTROLES
# ==============================================================================

case_select = Select(
    title="Caso",
    value=initial_case,
    options=list(cases.keys()),
    width=260
)

period_slider = Slider(
    title="Desplazamiento entero N",
    start=1,
    end=20,
    value=N0,
    step=1,
    width=250
)

info = Div(width=650)

info.text = """
<div style="
    padding: 10px;
    border-left: 5px solid seagreen;
    background-color: #eef8f1;
">
    <b>Sinusoide discreta periódica</b><br><br>

    Para Ω₀ = π/4, el período fundamental es N₀ = 8.

    <br><br>

    Con N = 8:

    <br>

    <span style="font-size:18px;">
        x[n+8] = x[n].
    </span>
</div>
"""

# ==============================================================================
# 5. CALLBACK
# ==============================================================================

callback = CustomJS(
    args=dict(
        source=source,
        case_select=case_select,
        period_slider=period_slider,
        info=info
    ),
    code="""
    const data = source.data;
    const n = data["n"];

    const cases = {
        "Periódica: Ω₀ = π/4": {
            omega: Math.PI / 4,
            label: "π/4",
            periodic: true,
            period: 8
        },

        "Periódica: Ω₀ = 2π/5": {
            omega: 2 * Math.PI / 5,
            label: "2π/5",
            periodic: true,
            period: 5
        },

        "No periódica: Ω₀ = π/√2": {
            omega: Math.PI / Math.sqrt(2),
            label: "π/√2",
            periodic: false,
            period: null
        },

        "No periódica: Ω₀ = 2/5": {
            omega: 2 / 5,
            label: "2/5",
            periodic: false,
            period: null
        }
    };

    const selectedCase = cases[case_select.value];

    const omega = selectedCase.omega;
    const N = period_slider.value;

    const x = [];
    const xShift = [];
    const differences = [];

    for (let i = 0; i < n.length; i++) {
        const ni = n[i];

        const value = Math.cos(omega * ni);
        const shiftedValue = Math.cos(omega * (ni + N));

        x.push(value);
        xShift.push(shiftedValue);

        differences.push(
            Math.abs(value - shiftedValue)
        );
    }

    data["x"] = x;
    data["x_shift"] = xShift;

    const maxDifference = Math.max(...differences);
    const coincide = maxDifference < 1e-10;

    if (selectedCase.periodic) {
        const N0 = selectedCase.period;
        const isMultiple = N % N0 === 0;

        if (isMultiple && coincide) {
            info.text = `
            <div style="
                padding: 10px;
                border-left: 5px solid seagreen;
                background-color: #eef8f1;
            ">
                <b>Sinusoide discreta periódica</b><br><br>

                Para Ω₀ = ${selectedCase.label}, se cumple

                <br><br>

                <span style="font-size:18px;">
                    Ω₀/(2π) ∈ ℚ.
                </span>

                <br><br>

                Su período fundamental es

                <span style="font-size:17px;">
                    N₀ = ${N0}.
                </span>

                <br><br>

                Como N = ${N} es múltiplo de ${N0}:

                <br>

                <span style="font-size:18px;">
                    x[n+${N}] = x[n].
                </span>
            </div>
            `;
        }

        else {
            info.text = `
            <div style="
                padding: 10px;
                border-left: 5px solid darkorange;
                background-color: #fff8e8;
            ">
                <b>La señal es periódica, pero N = ${N} no es un período.</b>
                <br><br>

                Para Ω₀ = ${selectedCase.label}, el período fundamental es

                <span style="font-size:17px;">
                    N₀ = ${N0}.
                </span>

                <br><br>

                El desplazamiento seleccionado no es múltiplo de N₀:

                <br>

                <span style="font-size:18px;">
                    x[n+${N}] ≠ x[n].
                </span>

                <br><br>

                Prueba N = ${N0}, ${2 * N0}, ...
            </div>
            `;
        }
    }

    else {
        info.text = `
        <div style="
            padding: 10px;
            border-left: 5px solid firebrick;
            background-color: #fff1f1;
        ">
            <b>Sinusoide discreta no periódica</b><br><br>

            Para Ω₀ = ${selectedCase.label}, se cumple

            <br><br>

            <span style="font-size:18px;">
                Ω₀/(2π) ∉ ℚ.
            </span>

            <br><br>

            Por tanto, no existe ningún entero positivo N que permita
            escribir

            <br><br>

            <span style="font-size:18px;">
                Ω₀N = 2πk,
            </span>

            <br><br>

            con k entero.

            Para N = ${N}:

            <br>

            <span style="font-size:18px;">
                x[n+${N}] ≠ x[n].
            </span>

            <br><br>

            Ningún valor entero no nulo de N produce una repetición exacta
            de la secuencia.
        </div>
        `;
    }

    source.change.emit();
    """
)

case_select.js_on_change("value", callback)
period_slider.js_on_change("value", callback)

# ==============================================================================
# 6. AJUSTE AUTOMÁTICO DEL DESPLAZAMIENTO AL CAMBIAR DE CASO
# ==============================================================================

case_callback = CustomJS(
    args=dict(
        case_select=case_select,
        period_slider=period_slider
    ),
    code="""
    const defaultPeriods = {
        "Periódica: Ω₀ = π/4": 8,
        "Periódica: Ω₀ = 2π/5": 5,
        "No periódica: Ω₀ = π/√2": 8,
        "No periódica: Ω₀ = 2/5": 5
    };

    period_slider.value = defaultPeriods[case_select.value];
    """
)

case_select.js_on_change("value", case_callback)

# ==============================================================================
# 7. LAYOUT
# ==============================================================================

layout = column(
    p,
    row(case_select, period_slider),
    info
)

show(layout)
````

### Ambigüedad
\begin{equation}
  e^{j(\Omega_0+2\pi)n}=e^{j\Omega_0 n}\underbrace{e^{j2\pi n}}_1=e^{j\Omega_0 n}.
\end{equation}
Una exponencial compleja discreta de frecuencia $\Omega_0$ es la misma que una de frecuencia $\Omega_0+2\pi$, ..., $\Omega_0+2\pi k,\ k\in\Z$.

Al contrario del caso continuo, en el cual todas las exponenciales complejas son distintas para distintos valores de $\omega_0$, en el caso discreto **no son todas distintas**. Sólo es necesario considerar un intervalo para la frecuencia fundamental de longitud $2\pi$. Por conveniencia se usa:
\begin{equation}
  0\le\Omega_0<2\pi, \text{ ó } -\pi\le\Omega_0<\pi.
\end{equation}
Debido a esta periodicidad en la frecuencia fundamental, no hay un incremento continuo de la velocidad de oscilación con la frecuencia fundamental:
\begin{equation}
  \begin{cases}
    0<\Omega_0<\pi\ &\Rightarrow\ \text{Incremento de velocidad de oscilación,}\\
    \pi<\Omega_0<2\pi\ &\Rightarrow\ \text{Disminución de velocidad de oscilación.}
  \end{cases}
\end{equation}

```{figure} figures/T1/2_3_fig11 
---
width: 85.0%
---
```


Las exponenciales de baja frecuencia son las que tienen frecuencias cercanas a $0$ ($2\pi k,\ k\in\Z$). \
Las exponenciales de alta frecuencia son las que tienen frecuencias cercanas a $\pi$ ($\pi(2k+1),\ k\in\Z$).

En particular:
```{math}
e^{j\pi n}=(e^{j\pi})^n=(-1)^n,
```
cambia de signo en cada punto. Es la señal discreta de mayor frecuencia posible.


### Completitud
Está conectado con las exponenciales complejas relacionadas armónicamente.

Al igual que para el caso de señales continuas, tiene interés considerar **exponenciales complejas discretas periódicas relacionadas armónicamente** (con período común $N$):
\begin{equation}
  \phi_k[n]=e^{j\Omega_k n}.
\end{equation}
Todas las exponenciales complejas relacionadas armónicamente tienen la propiedad de que son periódicas con un período común $N$[^4]:
[^4]: Siguiente la notación del Oppenheim, ahora usamos para el período común la notación $N$ y no $N_0$.
\begin{equation}
  x[n]=x[n+N]\ \Rightarrow\ e^{j\Omega_k N}=1.
\end{equation}
\begin{equation}
  \Omega_k N=2\pi k,\ k\in\Z.
\end{equation}
Si definimos $\Omega_0$ como[^5]:
[^5]: $\Omega_0$ no es la frecuencia fundamental de cada una, que será $\Omega_k$.
\begin{equation}
  \Omega_0=\frac{2\pi}{N},
\end{equation}
la frecuencia fundamental de cada una de las exponenciales complejas relacionadas armónicamente será:
\begin{equation}
  \Omega_k=k\Omega_0=k\frac{2\pi}{N}.
\end{equation}
Sus frecuencias fundamentales, $\Omega_k$, son múltiplos enteros de una sola frecuencia, $\Omega_0>0$.
\begin{equation}
  \phi_k[n]=e^{jk\Omega_0 n}=e^{jk\frac{2\pi}{N} n}.
\end{equation}

Pues bien, en el caso continuo todas las exponenciales complejas relacionadas armónicamente con un cierto período $T_0$ son distintas, y hay infinitas:
\begin{equation}
  \phi_k(t)=e^{jk\frac{2\pi}{T_0}t},\ k=0, \pm 1, \pm 2, ...
\end{equation}
Para el caso discreto esto no es así, ya que:
\begin{equation}
  \phi_{k+N}[n]=e^{j(k+N)\frac{2\pi}{N}n}=e^{jk\frac{2\pi}{N}n}\underbrace{e^{j2\pi n}}_1=e^{jk\frac{2\pi}{N}n}=\phi_k[n].
\end{equation}
Por lo tanto, **sólo existen $N$ exponenciales complejas discretas relacionadas armónicamente de período $N$**.
\begin{equation}
  \phi_0[n]=\phi_N[n],\ \phi_1[n]=\phi_{N+1}[n],\ \phi_{-1}[n]=\phi_{N-1}[n],\ ...
\end{equation}

### Resumen
En la tabla se muestra un resumen comparativo de las propiedades de las exponenciales complejas senoidales continuas y discretas.

|                                              | **Continua** $t\in\mathbb{R},\ \omega_0 (rad/s)$                                                                                    | **Discreta** $n\in\mathbb{Z},\ \Omega_0 (rad/muestra)$                                                                                                            |
| -------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Periodicidad**                             | Siempre son periódicas. $ \\ (T_0=\dfrac{2\pi}{\omega_0})$.                                                                               | No todas las exponenciales complejas son periódicas. **Debe cumplirse: $\Omega_0 N=2\pi m,\ m\in\mathbb{Z}$.** $N_0$ es el mínimo $N$ que cumple esa relación.     |
| **Ambigüedad**                               | Son todas distintas.<br>$e^{j(\omega_0 + 2\pi k)t}=e^{j\omega_0 t},e^{j 2 \pi k t}\neq e^{j \omega_0 t}$.                              | Dos exponenciales son iguales si $\Omega_0$ varía en un múltiplo entero de $2\pi$. $e^{j(\Omega_0 + 2\pi k)n}=e^{j\Omega_0 n},e^{j 2\pi k n}=e^{j\Omega_0 n}$.    |
| **Combinación lineal de señales periódicas** | No siempre periódica. Ej.: $T_1=1,\ T_2=\sqrt{2}$ ⇒ no existe $T_0$ con $n_1T_1=n_2T_2$.                                            | Siempre periódica. Si $N_{1}, N_{2}$ son períodos ⇒ $N_0=\mathrm{mcm}(N_1,N_2)$ y $N_0=n_1N_1=n_2N_2$.                                                            |
| **Completitud**                              | Infinitas exponenciales complejas armónicas de período $T_0$. $\phi_k(t)=e^{j\omega_k t}=e^{jk\omega_0 t}=e^{jk\frac{2\pi}{T_0}t}$. | Sólo hay $N$ exponenciales complejas armónicas de período $N$. $\phi_{k}[n]=e^{j k \frac{2\pi}{N}n}$. $\phi_{k+N}[n]= e^{j(k+N)\frac{2\pi}{N}n}= \phi_{k}[n]$. |
| **Velocidad de variación**                   | Aumenta linealmente con la frecuencia. <br>![](figures/T1/2_3_fig16 )                                                                 | **Bajas Frecuencias**: $\Omega_0$ cercano a $0$. **Frecuencias altas**: $\Omega_0$ cercano  a $\pi$. ![](figures/T1/2_3_fig17 )                                                |
