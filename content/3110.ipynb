### Series de Fourier

:::{warning .simple icon=false} Ejercicio 1 
La señal periódica que se muestra a continuación se genera cuando una señal de tensión con forma de coseno es rectificada por un diodo, proceso que se conoce como *rectificación de media onda*. Escriba esta señal como una suma de senos y cosenos armónicamente relacionados.

```{figure} figures/T3/signal_cos_rect.svg
:name: figs:T3:SF:ejercicio1
:width: 90%
:alt: Rectificación de media onda de señal coseno.
:align: center
```
:::

:::{tip .simple icon=false class=dropdown} Resolución del Ejercicio 1

Lo que se pide realmente es la serie de Fourier de la señal $x(t)$. Por ser una señal real, se va a poder expresar como una suma de senos y cosenos de la forma:
$$
\tilde{x}(t)=C_0 + 2\sum_{k=1}^{\infty} a_k \cos\left(k \omega_0 t\right) + 2\sum_{k=1}^{\infty} b_k \sen\left(k \omega_0 t\right),
$$
donde $a_k$ es la parte real de los coeficientes de la serie de Fourier, $c_k$, y $b_k$ su parte imaginaria.
%
Calculando explícitamente estos coeficientes:
\begin{equation}
\begin{split}
c_k	&= \frac{1}{2\pi}\int_{-\pi/2}^{\pi/2} \cos(t) e^{-jtk}dt
=\frac{1}{2\pi}\int_{-\pi/2}^{\pi/2} \frac{e^{jt}+e^{-jt}}{2} e^{-jtk}dt\\
	&= \vdots\\
	&= \frac{\cos\left(\frac{\pi}{2}k\right)}{\pi(1-k^2)},\quad k\neq \pm 1.
\end{split}
\end{equation}
%
Para $k=\pm 1$ la solución es $c_k=1/4$. Dado que los $c_k$ son reales (esto sucede porque la señal original $x(t)$ es hermítica, o más concretamente par), $a_k=c_k$, con lo que:
$$
\tilde{x}(t)=\frac{1}{\pi}+2\sum_{k=1}^{\infty}c_k \cos(kt).
$$
:::

A continuación, se muestra una animación de la aproximación a la serie de Fourier del coseno rectificado añadiendo N armónicos.

````python
%load_ext autoreload
%autoreload 2

import sys
import os
sys.path.append(os.path.abspath(".."))

from utils.plot_helpers import style_math_axes
````

````python
import numpy as np
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, CustomJS, Slider, Div, Label
from bokeh.layouts import column
from bokeh.io import output_notebook

# Importamos las funciones auxiliares
from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook()

# ==============================================================================
# 1. DATOS PRE-CALCULADOS
# ==============================================================================
# Rango temporal de -3*pi a 3*pi
t = np.linspace(-3 * np.pi, 3 * np.pi, 3000)

# Señal ideal: Coseno rectificado de media onda
ideal_signal = np.maximum(0, np.cos(t))

def fourier_rect_cos(t, n_harmonics):
    # N=0: Componente continua (c_0 = a_0/2 = 1/pi)
    approx = np.ones_like(t) / np.pi
    
    if n_harmonics >= 1:
        # N=1: Armónico fundamental (a_1 = 1/2)
        approx += 0.5 * np.cos(t)
    
    # N >= 2: Armónicos pares (los impares > 1 son nulos por la simetría)
    for k in range(2, n_harmonics + 1, 2):
        coef = (2 / np.pi) * ((-1)**(k/2 + 1)) / (k**2 - 1)
        approx += coef * np.cos(k * t)
        
    return approx

# Incrementos de 1 en 1, desde N=0 hasta N=50
N_values = list(range(0, 51))

data_dict = {'t': t, 'ideal': ideal_signal}

for n in N_values:
    data_dict[f'y_{n}'] = fourier_rect_cos(t, n)

# Empezamos con N=0
source = ColumnDataSource(data=dict(t=t, ideal=ideal_signal, y=data_dict['y_0']))

# ==============================================================================
# 2. CONFIGURACIÓN VISUAL
# ==============================================================================
COLOR_APROX = 'blue'

X_DATA_RANGE = (-3 * np.pi, 3 * np.pi)
Y_DATA_RANGE = (-0.3, 1.4)

p_main = figure(height=400, width=800, tools="pan,wheel_zoom,reset",
                title="Aproximación de la serie de Fourier (Coseno Rectificado)")

style_math_axes(p_main, 
                x_range=X_DATA_RANGE, 
                y_range=Y_DATA_RANGE, 
                xlabel="t", 
                ylabel=r"$$\tilde{x}_N(t)$$")

# Ticks en múltiplos de pi/2 con formato fraccionario en LaTeX
pi_ticks = (np.arange(-6, 7) * np.pi / 2).tolist()
pi_labels = [
    r"$$-3\pi$$", r"$$-\frac{5\pi}{2}$$", r"$$-2\pi$$", r"$$-\frac{3\pi}{2}$$",
    r"$$-\pi$$", r"$$-\frac{\pi}{2}$$", r"$$0$$", r"$$\frac{\pi}{2}$$",
    r"$$\pi$$", r"$$\frac{3\pi}{2}$$", r"$$2\pi$$", r"$$\frac{5\pi}{2}$$", r"$$3\pi$$"
]
add_math_ticks(p_main, xticks=pi_ticks, xtick_labels=pi_labels, yticks=[1], ytick_labels=["1"], tick_len=5, font_size="8pt")

# Fórmula Dinámica
N_init = N_values[0]
if N_init == 0:
    latex_text_init = r"$$\tilde{x}_0(t) = \frac{1}{\pi}$$"
else:
    latex_text_init = r"$$\tilde{x}_{" + str(N_init) + r"}(t) = \frac{1}{\pi} + \sum_{k=1}^{" + str(N_init) + r"} c_k \cos(k t)$$"
formula_label = Label(x=-2.5*np.pi, y=1.1, text=latex_text_init, text_font_size="11pt", text_color="#333")
p_main.add_layout(formula_label)

# Curvas
p_main.line('t', 'ideal', source=source, color="black", alpha=0.3, line_width=2, line_dash="dashed", legend_label="Ideal")
p_main.line('t', 'y', source=source, color=COLOR_APROX, line_width=2.5, legend_label="Aprox")
p_main.varea(x='t', y1='ideal', y2='y', source=source, fill_color="red", fill_alpha=0.2, legend_label="Error")

p_main.legend.location = "top_right"
p_main.legend.click_policy = "hide"
p_main.legend.border_line_color = None

# ==============================================================================
# 3. INTERACTIVIDAD
# ==============================================================================
div_info = Div(text=f"<h3 style='margin-bottom:0px; color:#333'>Armónicos: N = {N_values[0]}</h3>", sizing_mode="stretch_width")

# Slider configurado de 1 en 1 y ocupando todo el ancho
slider = Slider(start=0, end=len(N_values)-1, value=0, step=1, title="Desliza para aumentar N", show_value=False, sizing_mode="stretch_width") 

callback = CustomJS(args=dict(source=source, slider=slider, div=div_info, 
                              all_data=data_dict, N_vals=N_values, formula_label=formula_label), code="""
    const idx = slider.value;
    const N = N_vals[idx];
    const data = source.data;
    const y_new = all_data['y_' + N];
    
    for (let i = 0; i < data['y'].length; i++) { 
        data['y'][i] = y_new[i]; 
    }
    
    div.text = "<h3 style='margin-bottom:0px; color:#333'>Armónicos: N = " + N + "</h3>";
    if (N === 0) {
        formula_label.text = "$$\\\\tilde{x}_0(t) = \\\\frac{1}{\\\\pi}$$";
    } else {
        formula_label.text = "$$\\\\tilde{x}_{" + N + "}(t) = \\\\frac{1}{\\\\pi} + \\\\sum_{k=1}^{" + N + "} c_k \\\\cos(k t)$$";
    }
    source.change.emit();
""")

slider.js_on_change('value', callback)

texto_caption = """
<div style="font-family: sans-serif; margin-top: 10px; font-size: 14px; opacity: 0.8;">
    <b>Convergencia uniforme:</b> A diferencia de la onda cuadrada, esta señal es <b>continua</b> en el tiempo (no tiene saltos, solo "picos" en su derivada). 
    Por ello, <b>NO presenta Efecto de Gibbs</b>. Observa cómo el error (área roja) simplemente se aplana sin generar sobreimpulsos en los bordes.
</div>
"""
caption = Div(text=texto_caption, sizing_mode="stretch_width")

layout_final = column(div_info, slider, p_main, caption, sizing_mode="stretch_width")
show(layout_final)
````

:::{tip .simple icon=false class=dropdown} Resolución del Ejercicio 2

La señal $\tilde{x}(t)$ es periódica de periodo fundamental $T_0=2$. Tal y como está definida, la señal no cumple el criterio de convergencia de las series de Fourier:
$$
\int_{\langle T_0 \rangle}|x(t)|^2dt< \infty.
$$
Sin embargo, sí podremos calcular su desarrollo en serie de Fourier en términos de su periodo fundamental $T_0$ y los coeficientes $c_k$, siendo éste de la forma:
$$
\tilde{x}(t)=\sum_{k=-\infty}^{\infty}c_k e^{jk\pi t}.
$$
Los coeficientes se calculan en este caso a partir de la definición genérica:
$$
c_k=\frac{1}{2}\int_{\langle T_0 \rangle} \tilde{x}(t) e^{-jk\pi t}dt.
$$
Tomamos el periodo de integración evitando que sus extremos coincidan con las propias deltas, por ejemplo $[-1/2,3/2)$ (esto no es estrictamente necesario, pero simplifica los cálculos):
\begin{equation}
    \begin{split}
    c_k &= \frac{1}{2}\int_{-1/2}^{3/2} \left[ \delta(t)-2\delta(t-1) \right] e^{-jk\pi t}dt\\
        &= \frac{1}{2}\int_{-1/2}^{3/2} \delta(t) e^{-jk\pi t}dt -\int_{-1/2}^{3/2} \delta(t-1) e^{-jk\pi t}dt \\
        &= \frac{1}{2}\int_{-1/2}^{3/2} \delta(t) e^{0}dt -\int_{-1/2}^{3/2} \delta(t-1) e^{-jk\pi}dt \\
        &= \frac{1}{2} - e^{-jk\pi}=\frac{1}{2} - (-1)^{k}.
    \end{split}
\end{equation}
Con lo que $c_k=\frac{1}{2} - (-1)^{k}$, $k\in \Z$.
